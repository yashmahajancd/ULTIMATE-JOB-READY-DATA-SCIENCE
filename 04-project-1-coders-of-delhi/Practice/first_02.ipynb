```python
import json
```

```python
def load_data(filename):
    with open(filename, "r") as f:
        data = json.load(f)
    return data
```

```python
data = load_data("data.json")
```

```python
def display_data(data):
    print("Users name with thier connections:")
    print("--------------------------------------------------------------------")
    for user in data['users']:
        print(f"ID:{user['id']} - {user['name']} with friends are {user['friends']} and liked pages are {user['liked_pages']}")
    print("\nPages Information:")
    print("------------------------------------")
    for page in data['pages']:
        print(f"{page['id']} - {page['name']}")

display_data(data)
```
