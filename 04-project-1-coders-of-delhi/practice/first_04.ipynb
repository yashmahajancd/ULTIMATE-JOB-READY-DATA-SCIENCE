```python
import json

# write a function to load the data
def load_data(filename):
    with open(filename, "r") as file:
        return json.load(file)

# write a function to display users and their connections
def display_users(data):
    print("Users and their connections:")
    print("--------------------------------------------------------------")
    for user in data['users']:
        print(f"ID:{user['id']} - {user['name']} are friends with {user['friends']} and liked pages are {user['liked_pages']}")

    print("\nPages Information:")
    print("--------------------------------------")
    for page in data['pages']:
        print(f"{page['id']} - {page['name']}")

# load the data
data = load_data("cleaned_data5.json")
display_users(data)
```
