# Data Collection
You hire Vijyalaxmi Iyer and Sam Chandra who are from Hebbal and HSR Layout respectively

HSR layout and Hebbal are far away so you set up a meeting at Rameshwaram cafe, Indiranagar

```python
with open("finaldata.txt", encoding='utf-8') as file:
    data = file.read()
```

```python
chunks = data.split("\n\n")
chunks = [c for c in chunks if len(c) > 3]
```

```python
def parse_chunk(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split('\n')
    username = sep_chunk[0]
    no_of_posts = int(sep_chunk[1].split(" ")[0].replace(",", ""))
    no_of_followers = float(sep_chunk[2].split(" ")[0].replace(",", "").replace("K", "").replace("M", ""))
    if("K" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000)
    elif("M" in sep_chunk[2]):
        no_of_followers = int(no_of_followers * 1000000)
    else:
        no_of_followers = int(no_of_followers)
        
    no_of_following = float(sep_chunk[3].split(" ")[0].replace(",", "").replace("K", "").replace("M", ""))
    if("K" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000)
    elif("M" in sep_chunk[3]):
        no_of_following = int(no_of_following * 1000000)
    else:
        no_of_following = int(no_of_following)
    
    name = sep_chunk[4]
    if(len(sep_chunk) > 5):
        type_of_page = sep_chunk[5]
        bio = "\n".join(sep_chunk[6:])
    else:
        type_of_page = "Unknown"
        bio = ""
    # print(username, no_of_posts, no_of_followers, no_of_following, name, type_of_page, bio, sep=("\n"))
    return {"username": username, "no_of_posts": no_of_posts, "no_of_followers": no_of_followers, 
            "no_of_following": no_of_following, "name": name, "type_of_page": type_of_page, "bio": bio}

# print(parse_chunk(chunks[0]))
```

```python
all_chunks = []

for chunk in chunks:
    parsed_chunk = parse_chunk(chunk)
    all_chunks.append(parsed_chunk)
```

```python
import json
```

```python
s = json.dumps(all_chunks, indent=4)

with open("data.json", "w") as f:
    f.write(s)
```

# Who has the maximum posts?
Lets write code to find max posts

```python
max = 0

for chunk in all_chunks:
    if(max < chunk['no_of_posts']):
        max = chunk['no_of_posts']
        chunk_with_max_post = chunk
        
# print(chunk_with_max_post)
print(chunk_with_max_post['username'], f"Posts : {chunk_with_max_post['no_of_posts']}", sep=("\n"))
```

# Who has maximum followers?
Lets write code to find max followers

```python
maxx = 0

for chunk in all_chunks:
    if(maxx < chunk['no_of_followers']):
        maxx = chunk['no_of_followers']
        chunk_with_max_followers = chunk

print(chunk_with_max_followers['username'], f"Followers : {chunk_with_max_followers['no_of_followers']}", sep=("\n"))
```
