```python
import numpy as np
```

```python
arr = np.array([1, 2, 3, 4, 5])
result = arr ** 2   # vectorized operation
print(result)
```

### Broadcasting: Scaling Arrays Without Extra Memory

```python
print(arr)
```

```python
result2 = arr + 10
```

```python
print(result2)
```

### Broadcasting with Arrays of Different Shapes

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([10, 20, 30])

res = arr1 + arr2
print(res)
```
