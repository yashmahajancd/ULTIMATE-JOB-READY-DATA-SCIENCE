```python
import pandas as pd
```

```python
tweets = pd.DataFrame({
    "tweet_id": [1, 2],
    "content": ["Let us Code", "More than fifteen chars are here!"]
})
```

```python
tweets
```

---

```python
def invalid_tweets(tweets):
    invalid_tweets_df = tweets[tweets['content'].str.len() > 15]
    return invalid_tweets_df[['content']]
```

```python
invalid_tweets(tweets)
```
