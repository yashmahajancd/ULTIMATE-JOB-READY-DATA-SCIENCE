```python
import pandas as pd
```

```python
patients = pd.DataFrame({
    "patient_id": [1, 2, 3, 4, 5],
    "patient_name": ["Daniel", "Alice", "Bob", "George", "Alain"],
    "conditions": ["YFEV COUGH", "", "DIAB100 MYOP", "ACNE DIAB100", "DIAB201"]
})
```

```python
patients
```

---

```python
def find_patients(patients):
    find_patients_df = patients[patients['conditions'].str.contains(r'\bDIAB1')]
    return find_patients_df
```

```python
find_patients(patients)
```
