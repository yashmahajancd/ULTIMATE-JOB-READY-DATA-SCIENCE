```python
import pandas as pd
```

```python
users = pd.DataFrame({
    "user_id": [1, 2],
    "name": ["aLice", "bOB"]
})
```

```python
users
```

---

```python
def fix_names(users):
    users["name"] = users["name"].str.capitalize()
    return users.sort_values(by="user_id")
```

```python
fix_names(users)
```
