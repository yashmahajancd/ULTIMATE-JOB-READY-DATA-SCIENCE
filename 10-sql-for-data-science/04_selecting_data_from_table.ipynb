# 📘 How to Select Data in MySQL

When working with databases, one of the most common tasks is retrieving data from tables. In MySQL, this is done using the `SELECT` statement.

This section explains:
- How to retrieve data using `SELECT`
- How to filter data using `WHERE`
- How comparison operators work
- How to handle `NULL` values
- How to sort and limit results
- How to use wildcards with `LIKE`

---

# 1️⃣ Basic SELECT Statement

To retrieve data from a table, MySQL uses the `SELECT` statement.

## Retrieve All Data from a Table

```sql
SELECT * FROM student;
```

### Explanation

- `SELECT` tells MySQL that we want to retrieve data.
- `*` means “all columns”.
- `FROM student` specifies the table name.

This query returns all rows and all columns from the `student` table.

---

## Retrieve Specific Columns

Instead of retrieving every column, you can choose only the columns you need.

```sql
SELECT name, grade FROM student;
```

### Explanation

This query returns only:
- `name`
- `grade`

from the `student` table.

This is useful when:
- You don’t need all columns
- You want cleaner output
- You want better performance

---

# 2️⃣ Using the WHERE Clause

The `WHERE` clause is used to filter rows based on a condition.

It allows MySQL to return only those records that match the specified condition.

---

## Example: Students in 10th Grade

```sql
SELECT * FROM student
WHERE grade = '10th';
```

### Explanation

We are telling MySQL:

> “Return only those students whose grade is `10th`.”

---

## Example: Students Older Than 16

```sql
SELECT * FROM student
WHERE age > 16;
```

### Explanation

This query returns students whose age is greater than 16.

---

# 3️⃣ Comparison Operators in MySQL

Comparison operators are used inside conditions to compare values.

| Operator | Description | Example |
|----------|-------------|----------|
| `=` | Equal to | `WHERE age = 16` |
| `!=` | Not equal to | `WHERE grade != '12th'` |
| `<>` | Alternative for not equal to | `WHERE grade <> '12th'` |
| `>` | Greater than | `WHERE age > 16` |
| `<` | Less than | `WHERE age < 17` |
| `>=` | Greater than or equal to | `WHERE age >= 16` |
| `<=` | Less than or equal to | `WHERE age <= 18` |
| `BETWEEN` | Value within a range | `WHERE age BETWEEN 15 AND 17` |
| `IN` | Matches values from a list | `WHERE grade IN ('10th', '12th')` |
| `NOT IN` | Excludes list values | `WHERE grade NOT IN ('9th', '11th')` |
| `LIKE` | Pattern matching | `WHERE name LIKE 'A%'` |
| `NOT LIKE` | Opposite of LIKE | `WHERE name NOT LIKE '%a'` |

---

# 4️⃣ Handling NULL Values

## What is NULL?

`NULL` represents:
- missing values
- unknown values
- undefined data

It is important to understand that `NULL` is not:
- zero (`0`)
- an empty string (`''`)
- FALSE

It simply means:
> “No value exists.”

---

# ❌ Common Mistake While Checking NULL

```sql
SELECT * FROM student
WHERE grade = NULL;
```

This query will **not work correctly**.

Why?

Because `NULL` cannot be compared using normal comparison operators like `=` or `!=`.

---

# ✅ Correct Way to Check NULL Values

MySQL provides special operators for checking `NULL`.

| Condition | Correct Syntax |
|-----------|----------------|
| Check NULL values | `WHERE grade IS NULL` |
| Check NOT NULL values | `WHERE grade IS NOT NULL` |

---

## Example: Select Students with NULL Grades

```sql
SELECT * FROM student
WHERE grade IS NULL;
```

### Explanation

We are telling MySQL:

> “Return students whose grade value is missing.”

---

## Example: Select Students with Existing Grades

```sql
SELECT * FROM student
WHERE grade IS NOT NULL;
```

### Explanation

This query returns only those students whose grade contains a valid value.

---

# 5️⃣ Combining Conditions

MySQL allows multiple conditions using:
- `AND`
- `OR`
- parentheses `()`

This helps create more complex filtering logic.

---

## Example Using AND

```sql
SELECT * FROM student
WHERE grade = '10th' AND age > 16;
```

### Explanation

We are telling MySQL:

> “Return students who are in 10th grade AND older than 16.”

Both conditions must be true.

---

## Example Using OR

```sql
SELECT * FROM student
WHERE grade = '9th' OR grade = '12th';
```

### Explanation

This query returns students who are either:
- in 9th grade
- OR in 12th grade

Only one condition needs to be true.

---

## Example Using Parentheses

```sql
SELECT * FROM student
WHERE (grade = '10th' OR grade = '11th')
AND age >= 16;
```

### Explanation

This query:
1. First checks students in grades 10th or 11th
2. Then filters those whose age is 16 or more

Parentheses help control condition order.

---

# 6️⃣ Sorting Results with ORDER BY

The `ORDER BY` clause is used to sort query results.

Sorting can be:
- Ascending (`ASC`)
- Descending (`DESC`)

---

## Sort by Age in Ascending Order

```sql
SELECT * FROM student
ORDER BY age ASC;
```

### Explanation

This sorts students from:
- smallest age → largest age

`ASC` means ascending order.

---

## Sort by Name in Descending Order

```sql
SELECT * FROM student
ORDER BY name DESC;
```

### Explanation

This sorts names in reverse alphabetical order.

`DESC` means descending order.

---

# 7️⃣ Limiting Results with LIMIT

The `LIMIT` clause restricts how many rows MySQL returns.

This is useful when:
- previewing data
- pagination
- testing queries

---

## Example: Return Only 5 Rows

```sql
SELECT * FROM student
LIMIT 5;
```

### Explanation

This query returns only the first 5 rows.

---

## Example: Using LIMIT with Offset

```sql
SELECT * FROM student
LIMIT 2, 5;
```

### Explanation

This means:
- Skip first 2 rows
- Then return next 5 rows

---

# 8️⃣ Using Wildcards with LIKE

Wildcards are used with the `LIKE` operator to search patterns.

They are helpful when:
- you don’t know the exact value
- you want partial matching
- you want pattern-based searching

---

# Wildcards in MySQL

There are two main wildcards:

| Wildcard | Meaning |
|----------|----------|
| `%` | Matches zero or more characters |
| `_` | Matches exactly one character |

---

## Example: Names Starting with A

```sql
SELECT * FROM students
WHERE name LIKE 'A%';
```

### Explanation

This query returns names that start with `A`.

Examples:
- Aakash
- Aryan
- Ananya

The `%` wildcard means:
> “Any number of characters after A.”

---

# 📅 Matching Dates with `_` Wildcard

The `_` wildcard is useful for matching fixed-length patterns.

This is especially useful with dates stored in:

```text
YYYY-MM-DD
```

format.

---

## Example: Records from the 5th Day of Any Month

```sql
SELECT * FROM attendance
WHERE date LIKE '____-__-05';
```

---

## Explanation of Pattern

| Pattern Part | Meaning |
|--------------|----------|
| `____` | Any 4-digit year |
| `__` | Any 2-digit month |
| `05` | Day = 05 |

We are telling MySQL:

> “Return all rows where the date falls on the 5th day of any month and any year.”

---

# 📌 More Date Pattern Examples

| Pattern | Matches |
|----------|----------|
| `'2025-05-%'` | Any day in May 2025 |
| `'2024-12-__'` | All days in December 2024 |
| `'____-01-01'` | 1st January of any year |
| `'202_-__-__'` | Any date in the 2020s |
| `'____-__-3_'` | Dates with days from 30–39 |

---

# 🔁 Quick Recap: LIKE Wildcards

| Pattern | Meaning |
|----------|----------|
| `'A%'` | Starts with A |
| `'%sh'` | Ends with sh |
| `'%ar%'` | Contains “ar” |
| `'R____'` | 5-letter word starting with R |
| `'____-__-05'` | Dates where day = 05 |

---

# ✅ Summary

- `SELECT` retrieves data from tables.
- `WHERE` filters records.
- Comparison operators help create conditions.
- `NULL` must be checked using `IS NULL`.
- `ORDER BY` sorts results.
- `LIMIT` controls row count.
- `LIKE` helps perform pattern matching using wildcards.
