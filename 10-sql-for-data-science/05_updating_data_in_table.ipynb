# 📘 How to Update Data in a MySQL Table

In MySQL, updating data means modifying existing records in a table.  
This is done using the `UPDATE` statement.

The `UPDATE` statement allows you to:
- change one or more column values
- update specific rows using conditions
- modify existing information without deleting data

This operation is very important in real-world applications where data changes frequently, such as:
- updating customer information
- changing employee salaries
- correcting mistakes in records
- updating stock quantities

---

# 1️⃣ Basic UPDATE Syntax

The basic syntax of the `UPDATE` statement is:

```sql
UPDATE table_name
SET column1 = value1,
    column2 = value2
WHERE condition;
```

---

# 📖 Explanation of Syntax

| Keyword | Purpose |
|----------|----------|
| `UPDATE` | Specifies the table to modify |
| `SET` | Specifies the columns and new values |
| `WHERE` | Specifies which rows should be updated |

---

> ⚠️ Important:  
> If you omit the `WHERE` clause, MySQL will update **all rows** in the table.

---

# 2️⃣ Updating a Specific Row

Suppose we have a `student` table and we want to update the grade of a specific student.

## Example

```sql
UPDATE student
SET grade = '11th'
WHERE id = 1;
```

---

# 📖 Explanation

We are telling MySQL:

> “Find the student whose `id` is `1` and change the grade to `11th`.”

Only the matching row will be updated.

---

# 3️⃣ Updating Multiple Columns

You can update more than one column at the same time.

## Example

```sql
UPDATE student
SET age = 17,
    grade = '12th'
WHERE id = 2;
```

---

# 📖 Explanation

This query updates:
- `age`
- `grade`

for the student whose `id` is `2`.

---

# 4️⃣ Updating Multiple Rows

If the condition matches multiple rows, MySQL updates all matching rows.

## Example

```sql
UPDATE student
SET grade = 'Graduated'
WHERE age >= 18;
```

---

# 📖 Explanation

We are telling MySQL:

> “Update the grade to `Graduated` for all students whose age is 18 or more.”

All matching rows will be updated together.

---

# 5️⃣ Updating All Rows in a Table

If you do not use a `WHERE` clause, every row in the table gets updated.

## Example

```sql
UPDATE student
SET grade = 'Updated';
```

---

# ⚠️ Warning

This query changes the `grade` value for **every student** in the table.

Use this carefully because it affects all records.

---

# 6️⃣ Using Arithmetic Operations in UPDATE

MySQL also allows calculations while updating values.

## Example: Increase Age by 1

```sql
UPDATE student
SET age = age + 1;
```

---

# 📖 Explanation

Instead of assigning a fixed value, this query increases the existing `age` value by `1` for every row.

This is useful for:
- salary increments
- stock updates
- counters
- yearly updates

---

# 7️⃣ Verifying Updated Data

After updating records, you can check the table using `SELECT`.

## Example

```sql
SELECT * FROM student;
```

This displays all updated records from the table.

---

# ✅ Best Practices While Using UPDATE

- Always use a `WHERE` clause unless you intentionally want to update all rows.
- Verify conditions carefully before executing the query.
- Use `SELECT` first to preview affected rows.
- Backup important data before performing large updates.

---

# 🎯 Summary

- `UPDATE` is used to modify existing data.
- `SET` specifies new values.
- `WHERE` controls which rows are updated.
- Multiple columns can be updated together.
- Omitting `WHERE` updates every row.
- Arithmetic operations can also be used during updates.
