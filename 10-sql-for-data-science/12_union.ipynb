# 🔄 Using `UNION` in MySQL

The `UNION` operator is used to **combine the result sets of two or more `SELECT` statements** into a single result.

---

# 📌 Why Use `UNION`?

It helps when:

- You’re pulling **similar data** from different tables
- You want to **merge results** from multiple queries into one list

---

# ✅ When `UNION` Works

For `UNION` to work correctly:

1. All `SELECT` statements must have the **same number of columns**
2. Corresponding columns must have **compatible data types**
3. Columns are matched **by position**, not by name

---

## 💡 Example

```sql
SELECT name, city
FROM customers

UNION

SELECT name, city
FROM vendors;
```

---

## ✅ Result

This combines names and cities from both tables into a single result.

---

# ❌ When `UNION` Doesn’t Work

## 🔹 Different Number of Columns

```sql
-- This will throw an error

SELECT name, city
FROM customers

UNION

SELECT name
FROM vendors;
```

---

## 🔹 Different Data Types

```sql
-- Error if 'age' is an integer and 'name' is a string

SELECT age
FROM users

UNION

SELECT name
FROM students;
```

---

## ⚠️ Result

MySQL will complain because the columns cannot be matched due to a data type mismatch.

---

# 🔍 `UNION` vs `UNION ALL`

By default, `UNION` removes **duplicate rows**.

If you want to **keep duplicates**, use `UNION ALL`.

---

## 💡 Example Using `UNION ALL`

```sql
SELECT name
FROM students

UNION ALL

SELECT name
FROM alumni;
```

---

# 📌 When to Use Which?

## ✅ Use `UNION` When:

- You want a **clean list** without duplicates

---

## ✅ Use `UNION ALL` When:

- You want **better performance**
- You don’t care about duplicates
- You expect duplicate values and want to preserve them

---

# 💼 Practical Use Case Example

Let’s say you have two tables:

- `students_2023`
- `students_2024`

```sql
SELECT name, batch
FROM students_2023

UNION

SELECT name, batch
FROM students_2024;
```

---

## ✅ Result

This gives a combined list of all students across both years without duplicates.

---

# 📊 Sorting the Combined Result

You can sort the final output using `ORDER BY` at the end.

```sql
SELECT name
FROM students_2023

UNION

SELECT name
FROM students_2024

ORDER BY name;
```

---

# 📋 Summary

| Feature | Description |
|---|---|
| `UNION` | Combines results and removes duplicates |
| `UNION ALL` | Combines results and keeps duplicates |
| Requirements | Same column count and compatible data types |
| `ORDER BY` | Used at the end to sort final output |

---

# ✅ Conclusion

The `UNION` operator is useful when combining similar data from multiple tables or queries into a single result set.

It helps:

- Merge records efficiently
- Create cleaner reports
- Combine yearly or departmental data
- Simplify query results
