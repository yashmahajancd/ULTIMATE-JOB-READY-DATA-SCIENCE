# 👁️ MySQL Views

A **View** in MySQL is like a **virtual table**.

It doesn’t store data by itself but instead shows data from one or more tables through a saved SQL query.

You can use a view just like a regular table:

- `SELECT` from it
- Apply filters
- Join it with other tables
- Reuse it in queries

---

# 📌 Why Use Views?

Views are useful because they help:

- Simplify complex queries by giving them a name
- Hide sensitive columns from users
- Show only specific rows or columns from a table
- Reuse common query logic across applications or reports

---

# 🛠️ Creating a View

Let’s say you have an `employees` table with many details, but you only want to show public employee information.

```sql
CREATE VIEW public_employees AS

SELECT
    name,
    department,
    salary
FROM employees;
```

---

## 💡 What This Means

You’re telling MySQL:

> “Create a view called `public_employees`
> that shows only `name`, `department`, and `salary`
> from the `employees` table.”

---

# 📋 Using a View

Now you can query the view just like a normal table.

```sql
SELECT *
FROM public_employees;
```

---

## 🔍 Applying Filters

```sql
SELECT *
FROM public_employees
WHERE department = 'IT';
```

---

# 🔄 Updating a View

You can replace an existing view using:

```sql
CREATE OR REPLACE VIEW public_employees AS

SELECT
    name,
    department
FROM employees;
```

---

# ❌ Dropping (Deleting) a View

```sql
DROP VIEW public_employees;
```

This removes the view from the database.

---

# 📝 Important Notes

- Views do not store data physically
- If the underlying table changes, the view reflects those changes automatically
- Not all views are updatable

---

## ⚠️ Updatable vs Non-Updatable Views

Simple views usually allow:

- `INSERT`
- `UPDATE`
- `DELETE`

Complex views with:

- Joins
- Grouping
- Aggregate functions

may not allow updates.

---

# 💼 Example Use Case

Suppose this query is used many times across your application:

```sql
SELECT
    customer_id,
    name,
    total_orders,
    status
FROM customers
WHERE status = 'active'
    AND total_orders > 5;
```

Instead of repeating it everywhere, create a view.

---

# ✅ Create the View

```sql
CREATE VIEW top_customers AS

SELECT
    customer_id,
    name,
    total_orders,
    status
FROM customers
WHERE status = 'active'
    AND total_orders > 5;
```

---

# ✅ Use the View

```sql
SELECT *
FROM top_customers;
```

---

# 📋 Summary

| Feature | Description |
|---|---|
| View | A virtual table based on a query |
| Stores Data? | ❌ No |
| Uses | Simplify queries, security, reusable logic |
| Can Use `SELECT`? | ✅ Yes |
| Automatically Updated? | ✅ Yes |
| Can Be Deleted? | ✅ Using `DROP VIEW` |

---

# ✅ Conclusion

MySQL Views help make SQL queries cleaner, simpler, and easier to maintain.

They are especially useful for:

- Reusing complex queries
- Controlling data access
- Organizing database logic
- Improving readability

Views act like saved queries that behave just like tables.
