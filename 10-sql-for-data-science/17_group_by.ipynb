# 📊 `GROUP BY` in MySQL

The `GROUP BY` clause is used when you want to **group rows that have the same values in specified columns**.

It is usually combined with aggregate functions like:

- `COUNT()`
- `SUM()`
- `AVG()`
- `MAX()`
- `MIN()`

---

# 💡 What `GROUP BY` Means

We are telling MySQL:

> “Group these rows together by this column,
> and then apply an aggregate function to each group.”

---

# 🔹 Example: Count of Employees in Each Department

```sql
SELECT
    department,
    COUNT(*) AS total_employees
FROM employees
GROUP BY department;
```

---

## ✅ What This Does

We are grouping all employees by their department and counting how many employees are in each group.

---

# 🔹 Example: Average Salary Per Department

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employees
GROUP BY department;
```

---

## ✅ What This Does

We are telling MySQL:

> “Group the data by department,
> then calculate the average salary for each group.”

---

# 🔹 Using `GROUP BY` with Multiple Columns

You can group by more than one column for more detailed grouping.

---

# 💼 Example: Count by Department and Job Title

```sql
SELECT
    department,
    job_title,
    COUNT(*) AS count
FROM employees
GROUP BY department, job_title;
```

---

## ✅ What This Does

This counts how many employees hold each job title within each department.

---

# 🔍 The `HAVING` Clause

After grouping data using `GROUP BY`,
you may want to filter the groups themselves.

This is where `HAVING` is used.

---

# 📌 Important Difference

`HAVING` works **after grouping**,
while `WHERE` works **before grouping**.

---

# 💼 Example: Departments with More Than 5 Employees

```sql
SELECT
    department,
    COUNT(*) AS total
FROM employees
GROUP BY department
HAVING COUNT(*) > 5;
```

---

## ✅ What This Does

We are telling MySQL:

> “First group employees by department,
> then only show departments where the total number is greater than 5.”

---

# ⚖️ Difference Between `WHERE` and `HAVING`

| **Clause** | **Used For** | **Example Use** |
|---|---|---|
| `WHERE` | Filters rows before grouping | `WHERE salary > 50000` |
| `HAVING` | Filters groups after grouping | `HAVING AVG(salary) > 60000` |

---

# 🔹 Using `WHERE` and `HAVING` Together

```sql
SELECT
    department,
    AVG(salary) AS avg_salary
FROM employees
WHERE status = 'active'
GROUP BY department
HAVING AVG(salary) > 60000;
```

---

# ✅ What Happens Here

### Step 1 → `WHERE`

Filters only active employees.

---

### Step 2 → `GROUP BY`

Groups employees by department.

---

### Step 3 → `HAVING`

Filters departments with low average salary.

---

# 🔄 Using `WITH ROLLUP` in MySQL

The `WITH ROLLUP` clause is used with `GROUP BY`
to add:

- Summary rows
- Totals
- Subtotals

to the result set.

---

# 📋 Summary

| **Keyword** | **Role** |
|---|---|
| `GROUP BY` | Groups rows with same values into summary rows |
| `HAVING` | Filters grouped results based on aggregate values |
| `WITH ROLLUP` | Adds totals and subtotals |

---

# ✅ Conclusion

Use `GROUP BY` when you want to aggregate data.

Use `HAVING` when you want to filter those aggregated groups.

Together, they help you:

- Generate summaries
- Analyze grouped data
- Create reports
- Perform advanced data analysis
