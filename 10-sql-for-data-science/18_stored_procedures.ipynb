# ⚙️ Stored Procedures in MySQL

A **Stored Procedure** is a saved block of SQL code that you can execute later by calling its name.

It allows you to group SQL statements and reuse them — just like a function in programming.

---

# 📌 Why Use Stored Procedures?

Stored procedures are useful because they help:

- Avoid repeating the same SQL logic
- Improve performance by reducing network traffic
- Encapsulate complex business logic inside the database

---

# 🛠️ Creating a Stored Procedure

When creating a stored procedure, you usually need to temporarily change the SQL statement delimiter from `;` to another symbol like:

- `//`
- `$$`

---

# ❓ Why Change the `DELIMITER`?

MySQL normally ends a command at the first `;`.

Since stored procedures contain multiple SQL statements (each ending with `;`), we must tell MySQL:

> “Do not end the procedure too early.”

So we temporarily change the delimiter and later switch it back.

---

# 💼 Example: Simple Procedure to List All Employees

```sql
DELIMITER //

CREATE PROCEDURE list_employees()
BEGIN
    SELECT * FROM employees;
END //

DELIMITER ;
```

---

## ✅ What This Does

This creates a stored procedure named:

```sql
list_employees
```

---

# ▶️ Calling a Stored Procedure

Use the `CALL` statement to execute a procedure.

```sql
CALL list_employees();
```

---

# 🔹 Stored Procedure with Parameters

You can pass values into procedures using the `IN` keyword.

---

# 💼 Example: Get Employee Details by ID

```sql
DELIMITER //

CREATE PROCEDURE get_employee_by_id(IN emp_id INT)
BEGIN
    SELECT *
    FROM employees
    WHERE id = emp_id;
END //

DELIMITER ;
```

---

# 💡 What `IN emp_id INT` Means

We are telling MySQL:

> “Take an integer input called `emp_id`
> when this procedure is called.”

---

# ▶️ Call the Procedure

```sql
CALL get_employee_by_id(3);
```

---

# ❌ Dropping a Stored Procedure

To delete a stored procedure:

```sql
DROP PROCEDURE IF EXISTS get_employee_by_id;
```

---

## ✅ What This Does

This removes the procedure safely without throwing an error if it does not exist.

---

# 📋 Summary

| **Task** | **SQL Command** |
|---|---|
| Create Procedure | `CREATE PROCEDURE` |
| Change Delimiter | `DELIMITER //` *(or another symbol)* |
| Call a Procedure | `CALL procedure_name();` |
| Input Parameter | `IN param_name data_type` |
| Drop Procedure | `DROP PROCEDURE IF EXISTS procedure_name;` |

---

# ✅ Best Practices

- Give procedures clear and meaningful names
- Use:
  - `IN`
  - `OUT`
  - `INOUT`

  for flexible parameter handling

- Keep business logic in the database only when it improves clarity or performance

---

# 🔍 Additional Parameter Types

| **Type** | **Purpose** |
|---|---|
| `IN` | Pass value into the procedure |
| `OUT` | Return value from the procedure |
| `INOUT` | Pass and modify value |

---

# ✅ Conclusion

Stored Procedures help organize reusable SQL logic inside the database.

They are useful for:

- Reusability
- Performance optimization
- Cleaner application code
- Centralized business logic

Using stored procedures properly can make database applications easier to manage and maintain.
