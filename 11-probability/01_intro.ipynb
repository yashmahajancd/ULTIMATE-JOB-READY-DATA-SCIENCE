# 🎲 Introduction to Probability in Data Science

---

# 📌 1. What is Probability?

Probability is a measure of the likelihood that a particular event will occur.

It ranges from:

- `0` → Impossible event
- `1` → Certain event

Probability is used to model uncertainty with data and make informed decisions based on that uncertainty.

---

# 💡 Examples

- Tossing a coin
- Rolling a die
- Predicting customer churn

---

# 📌 2. Role of Probability in Data Science

Probability forms the foundation of:

- Statistical inference
- Machine learning
- Decision-making under uncertainty

---

# ✅ Why Data Scientists Use Probability

Data scientists use probability to:

- Estimate unknown values from sample data
- Evaluate risks and uncertainties
- Make predictions using probabilistic models

---

# 📌 3. Types of Probability

---

# 🔹 Theoretical Probability

Based on known possible outcomes.

### 💡 Example

Probability of getting heads in a fair coin:

```text
P(Heads) = 0.5
```

---

# 🔹 Empirical (Experimental) Probability

Based on observed data.

### 💡 Example

Probability that a user clicks an ad based on:

- 10,000 past impressions

---

# 📌 4. Key Concepts

---

# 🔹 Sample Space (S)

The set of all possible outcomes of an experiment.

---

# 🔹 Event (E)

A subset of the sample space containing one or more outcomes.

---

# 💼 Example: Rolling a Die

### 🎲 Sample Space

```text
S = {1, 2, 3, 4, 5, 6}
```

---

### 🎯 Event: Even Numbers

```text
E = {2, 4, 6}
```

# 🧮 Calculating Probability

---

# 📐 Formula

P(E) = Number of favorable outcomes / Total number of outcomes

---

# 📌 Where

- `P(E)` = Probability of event `E`
- Favorable outcomes = Outcomes satisfying the event
- Total outcomes = All possible outcomes in the sample space

---

# 💡 Example: Rolling a Die

If you roll a die, the probability of rolling a `3` is:

P(rolling a 3) = Number of favorable outcomes (1) / Total number of outcomes (6)
= 1/6 ≈ 0.1667
