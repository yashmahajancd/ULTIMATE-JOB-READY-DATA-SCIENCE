# 🎲 Types of Experiments

---

# 📌 1. Deterministic Experiment

A **deterministic experiment** always produces the same outcome when repeated under identical conditions.

---

## 💡 Example

- Adding `2 + 2` always gives `4`
- Heating water to `100°C` under normal pressure boils it

---

# 📌 2. Probabilistic / Random Experiment

A **probabilistic (random) experiment** has outcomes that are uncertain and may vary even when repeated under identical conditions.

---

## 💡 Example

- Tossing a coin
- Rolling a die
- Drawing a card from a deck

---

# 📚 Sample Space

The **sample space** is the set of all possible outcomes of a probabilistic experiment.

It is usually represented by:

```text
S
```

---

# 🎲 Examples of Sample Space

---

# 🔹 Rolling a Die

```text
S = {1, 2, 3, 4, 5, 6}
```

---

# 🔹 Tossing 2 Coins

```text
S = {(HH), (HT), (TH), (TT)}
```

---

# 🪙 For Coins

The number of possible outcomes for coins is:

Number of possible outcomes = 2ⁿ

Where:

- `n` = Number of coins

---

# 🎲 For Dice

The number of possible outcomes for dice is:

Number of possible outcomes = 6ⁿ

Where:

- `n` = Number of dice
