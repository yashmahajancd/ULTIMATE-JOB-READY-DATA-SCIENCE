# 1️⃣ Conditional Probability with Colored Balls

A box contains:

- 5 Red balls
- 3 Blue balls
- 2 Green balls

A ball is picked at random.

You are told that the ball picked is **not green**.

What is the probability that the ball is **red**?

# 2️⃣ Bayes’ Theorem – Football Players

A class consists of:

- 60% Boys
- 40% Girls

Among them:

- 30% of the boys play football
- 10% of the girls play football

If a randomly selected student is known to be a **football player**, what is the probability that the student is a **boy**?
