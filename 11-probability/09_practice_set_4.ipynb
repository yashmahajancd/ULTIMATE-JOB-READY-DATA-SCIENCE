# 2️⃣ Medical Diagnosis Using Bayes’ Theorem

- If a person has the disease, the test returns a positive result **99%** of the time.
- If a person does not have the disease, the test returns a positive result **5%** of the time.

If a randomly selected individual takes the test and obtains a **positive result**, what is the probability that the individual **actually has the disease**?
