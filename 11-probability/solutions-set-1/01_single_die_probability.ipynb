# 🎲 Question 1: Single Die Probability

**Problem:** You roll a fair six-sided die. What is the probability of getting an even number?

---

### 🧠 Mathematical Solution

When rolling a fair six-sided die, the sample space ($S$) consists of all possible outcomes:
$$S = \{1, 2, 3, 4, 5, 6\}$$
Total number of possible outcomes ($n(S)$) = 6

Let $E$ be the event of rolling an even number. The favorable outcomes in a die are:
$$E = \{2, 4, 6\}$$
Number of favorable outcomes ($n(E)$) = 3

The probability $P(E)$ is given by the formula:
$$P(E) = \frac{n(E)}{n(S)}$$

$$P(E) = \frac{3}{6} = \frac{1}{2}$$

> **Result:** The probability of getting an even number is **0.5** (or 50%).

```python
# Define the sample space and the event
sample_space = {1, 2, 3, 4, 5, 6}
even_numbers = {x for x in sample_space if x % 2 == 0}

# Calculate probabilities
total_outcomes = len(sample_space)
favorable_outcomes = len(even_numbers)

probability = favorable_outcomes / total_outcomes

# Display the results neatly
print(f"Sample Space: {sample_space}")
print(f"Favorable Outcomes (Even Numbers): {even_numbers}")
print(f"Probability of rolling an even number: {probability:.2f}")
```
