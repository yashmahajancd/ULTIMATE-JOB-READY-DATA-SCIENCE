# 🪙 Question 2: Two Coins Tossed

**Problem:** Two coins are tossed. Find the sample space associated with this random experiment.

---

### 🧠 Mathematical Solution

When a single coin is tossed, the possible outcomes are Head (H) or Tail (T). 
When **two** coins are tossed simultaneously, we find the Cartesian product of the two individual sample spaces.

Let $C_1 = \{H, T\}$ and $C_2 = \{H, T\}$.
The sample space $S$ for two coins is:
$$S = C_1 \times C_2 = \{(H, H), (H, T), (T, H), (T, T)\}$$

> **Result:** The sample space is **{HH, HT, TH, TT}**.

```python
import itertools

# Defining the outcomes for a single coin
coins = ['H', 'T']

# Generating sample space for 2 coins using Cartesian product
sample_space = list(itertools.product(coins, repeat=2))

# Formatting the output to match standard mathematical representation (e.g., 'HH')
sample_space_formatted = ["".join(outcome) for outcome in sample_space]

print("Experiment: Tossing 2 coins")
print("-" * 40)
print(f"Sample Space: {sample_space_formatted}")
print(f"Total Outcomes: {len(sample_space_formatted)}")
```
