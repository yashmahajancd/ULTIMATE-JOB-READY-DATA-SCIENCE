# 🎲🎲 Question 3: Sum of Two Dice

**Problem:** Two six-sided dice are rolled. What is the probability that the sum is 7?

---

### 🧠 Mathematical Solution

When two dice are rolled, the total number of possible combinations in the sample space ($S$) is:
$$n(S) = 6 \times 6 = 36$$

Let $E$ be the event where the sum of the numbers on the two dice equals 7. The favorable outcomes are:
$$E = \{(1, 6), (2, 5), (3, 4), (4, 3), (5, 2), (6, 1)\}$$
Number of favorable outcomes ($n(E)$) = 6

The probability $P(E)$ is:
$$P(E) = \frac{n(E)}{n(S)} = \frac{6}{36} = \frac{1}{6}$$

> **Result:** The probability that the sum is 7 is **1/6** (approx. 0.1667).

```python
# Generating sample space for two 6-sided dice
dice = [1, 2, 3, 4, 5, 6]
sample_space = [(d1, d2) for d1 in dice for d2 in dice]

# Finding favorable outcomes where the sum of the pair is exactly 7
favorable_outcomes = [outcome for outcome in sample_space if sum(outcome) == 7]

# Probability calculation
probability = len(favorable_outcomes) / len(sample_space)

print(f"Total possible outcomes: {len(sample_space)}")
print(f"Favorable outcomes (Sum = 7): {favorable_outcomes}")
print(f"Probability: {len(favorable_outcomes)}/{len(sample_space)} = {probability:.4f}")
```
