# 🔴🔵🟢 Question 4: Selecting a Red Ball

**Problem:** A bag contains 3 red, 2 blue, and 5 green balls. What is the probability of randomly selecting a red ball?

---

### 🧠 Mathematical Solution

First, we determine the total number of balls in the bag, which represents our sample space size ($n(S)$):
$$n(S) = 3 \text{ (red)} + 2 \text{ (blue)} + 5 \text{ (green)} = 10 \text{ balls}$$

Let $R$ be the event of drawing a red ball. The number of favorable outcomes is the total number of red balls:
$$n(R) = 3$$

The probability $P(R)$ is:
$$P(R) = \frac{n(R)}{n(S)} = \frac{3}{10}$$

> **Result:** The probability of randomly selecting a red ball is **0.3** (or 30%).

```python
# Defining the contents of the bag
bag = {
    'Red': 3,
    'Blue': 2,
    'Green': 5
}

# Calculations
total_balls = sum(bag.values())
red_balls = bag['Red']
probability_red = red_balls / total_balls

print(f"Bag contents: {bag}")
print(f"Total balls in the bag: {total_balls}")
print(f"Probability of drawing a Red ball: {red_balls}/{total_balls} = {probability_red:.2f}")
```
