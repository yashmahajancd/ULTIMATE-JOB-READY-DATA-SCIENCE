# 🃏 Question 5: Drawing from a Deck

**Problem:** A card is drawn at random from a deck. Find the probability of:
1. Getting a heart
2. Getting a face card

---

### 🧠 Mathematical Solution

A standard deck of playing cards contains **52 cards**. Therefore, our sample space size $n(S) = 52$.

**Part 1: Probability of getting a heart**
There are 4 suits in a deck, and each suit has 13 cards. 
Let $H$ be the event of drawing a heart.
$$n(H) = 13$$
$$P(H) = \frac{n(H)}{n(S)} = \frac{13}{52} = \frac{1}{4}$$

**Part 2: Probability of getting a face card**
The face cards are Jack (J), Queen (Q), and King (K). Since there are 4 suits, the total number of face cards is:
Let $F$ be the event of drawing a face card.
$$n(F) = 3 \text{ (face cards)} \times 4 \text{ (suits)} = 12$$
$$P(F) = \frac{n(F)}{n(S)} = \frac{12}{52} = \frac{3}{13}$$

> **Result:**
> - Probability of getting a heart: **0.25** (25%)
> - Probability of getting a face card: **$\approx 0.2307$** (23.07%)

```python
# Deck configuration constants
TOTAL_CARDS = 52
SUITS = 4

# Part 1: Getting a heart
hearts_in_deck = 13
prob_heart = hearts_in_deck / TOTAL_CARDS

# Part 2: Getting a face card (Jack, Queen, King for each suit)
face_cards_per_suit = 3
total_face_cards = face_cards_per_suit * SUITS
prob_face_card = total_face_cards / TOTAL_CARDS

print("--- 🃏 Card Probabilities ---")
print(f"Total cards in standard deck: {TOTAL_CARDS}\n")

print("1. Probability of getting a Heart:")
print(f"   Calculation: {hearts_in_deck}/{TOTAL_CARDS}")
print(f"   Decimal: {prob_heart:.2f}\n")

print("2. Probability of getting a Face Card:")
print(f"   Calculation: {total_face_cards}/{TOTAL_CARDS}")
print(f"   Decimal: {prob_face_card:.4f}")
```
