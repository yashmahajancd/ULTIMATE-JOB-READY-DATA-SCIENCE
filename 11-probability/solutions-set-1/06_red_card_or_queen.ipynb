# 🃏 Question 6: Probability of a Red Card or a Queen

**Problem:** A card is drawn from a standard deck. What is the probability of drawing a red card or a queen?

---

### 🧠 Mathematical Solution

This problem involves the **Addition Rule of Probability** for events that are not mutually exclusive (since a card can be both red and a queen at the same time).

The formula is:
$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

Where:
* Total cards in a standard deck ($n(S)$) = 52
* **Event A (Red Card):** There are 26 red cards (Hearts and Diamonds).
    $$P(A) = \frac{26}{52}$$
* **Event B (Queen):** There are 4 queens in a deck (one for each suit).
    $$P(B) = \frac{4}{52}$$
* **Event $A \cap B$ (Red Queen):** There are 2 red queens (Queen of Hearts and Queen of Diamonds).
    $$P(A \cap B) = \frac{2}{52}$$

### 🔢 Calculation

Now, substituting the values into our addition formula:
$$P(A \cup B) = \frac{26}{52} + \frac{4}{52} - \frac{2}{52}$$

$$P(A \cup B) = \frac{26 + 4 - 2}{52} = \frac{28}{52}$$

Reducing the fraction by dividing the numerator and denominator by 4:
$$P(A \cup B) = \frac{7}{13}$$

> **Result:** The probability of drawing a red card or a queen is **7/13** (approx. **0.5385** or 53.85%).

```python
# Initializing the standard deck parameters
total_cards = 52
red_cards = 26
total_queens = 4
red_queens = 2  # Intersection: cards that are both Red and Queen

# Calculating individual probabilities
p_red = red_cards / total_cards
p_queen = total_queens / total_cards
p_red_and_queen = red_queens / total_cards

# Applying the Addition Rule: P(A or B) = P(A) + P(B) - P(A and B)
p_red_or_queen = p_red + p_queen - p_red_and_queen

# Displaying output
print("--- Card Probability Results ---")
print(f"Probability of Red Card: {p_red:.4f}")
print(f"Probability of Queen: {p_queen:.4f}")
print(f"Probability of Red Queen (Overlap): {p_red_and_queen:.4f}")
print("-" * 55)
print(f"Probability of Red Card OR Queen: {p_red_or_queen:.4f} (or {p_red_or_queen * 100:.2f}%)")
```
