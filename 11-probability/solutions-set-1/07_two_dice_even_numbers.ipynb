# 🎲🎲 Question 7: Both Dice Showing Even Numbers

**Problem:** If two dice are rolled, what is the probability that both show even numbers?

---

### 🧠 Mathematical Solution

When two independent standard six-sided dice are rolled:

* Total outcomes in the sample space ($n(S)$):
    $$6 \times 6 = 36$$

Let $E$ be the event where **both** dice show even numbers. 
The even numbers on a single die are $\{2, 4, 6\}$ (3 possible choices).

* Favorable outcomes for the pair ($n(E)$) can be written as:
    $$E = \{(2,2), (2,4), (2,6), (4,2), (4,4), (4,6), (6,2), (6,4), (6,6)\}$$
    Total number of favorable outcomes = $3 \times 3 = 9$

### 🔢 Calculation

The probability $P(E)$ is:
$$P(E) = \frac{n(E)}{n(S)} = \frac{9}{36}$$

Simplifying the fraction:
$$P(E) = \frac{1}{4}$$

> **Alternative Approach (Independent Events):**
> Since the roll of the second die doesn't depend on the first, we can multiply their individual probabilities:
> $$P(\text{Even on Die 1}) = \frac{3}{6} = \frac{1}{2}$$
> $$P(\text{Even on Die 2}) = \frac{3}{6} = \frac{1}{2}$$
> $$P(\text{Both Even}) = \frac{1}{2} \times \frac{1}{2} = \frac{1}{4}$$

> **Result:** The probability that both dice show even numbers is **0.25** (or 25%).

```python
# Creating the sample space for two rolled dice
dice_faces = [1, 2, 3, 4, 5, 6]
sample_space = [(d1, d2) for d1 in dice_faces for d2 in dice_faces]

# Filtering outcomes where both values are even
favorable_outcomes = [outcome for outcome in sample_space if outcome[0] % 2 == 0 and outcome[1] % 2 == 0]

# Math operations
total_count = len(sample_space)
favorable_count = len(favorable_outcomes)
probability = favorable_count / total_count

print("--- Two Dice Roll Simulation ---")
print(f"Total possible combinations: {total_count}")
print(f"Favorable combinations (Both Even): {favorable_count}")
print(f"List of favorable pairs: {favorable_outcomes}")
print("-" * 100)
print(f"Probability: {favorable_count}/{total_count} = {probability:.2f} ({probability * 100:.1f}%)")
```
