# 🔄 Question 8: Dependent Events in Real Life

**Problem:** Think of an example from your daily life where two events are dependent.

---

### 🧠 Concept & Definition

Two events are **dependent** if the occurrence of the first event affects the probability of the occurrence of the second event. Mathematically, $P(B|A) \neq P(B)$.

### 🚗 Real-Life Example: Weather and Commute Times

Let's look at an everyday scenario that perfectly highlights dependency:
* **Event A:** It rains heavily during the morning rush hour.
* **Event B:** You reach your office or school late.

#### Why are they dependent?
On a regular clear day, the baseline probability of being late might be very low (e.g., 5%). However, if **Event A (Heavy Rain)** happens, it leads to waterlogged streets, slower driving speeds, and increased traffic accidents. This drastically spikes the probability of **Event B (Being Late)** to perhaps 60%. 

Since the occurrence of Event A alters the likelihood of Event B, these two events are clearly **dependent**.

| Scenario | Condition | Probability of Being Late ($P(\text{Late})$) |
| :--- | :--- | :--- |
| Clear Weather | Normal Conditions | Low ($\approx 5\%$) |
| Heavy Rain | Traffic Slowdowns | High ($\approx 60\%$) |

```python
import random

def simulate_commute(is_raining):
    """
    Simulates reaching late based on weather conditions to demonstrate dependency.
    """
    # Conditional probabilities showing dependency
    if is_raining:
        # High chance of being late due to traffic delays
        return random.choices(['Late', 'On Time'], weights=[0.60, 0.40])[0]
    else:
        # Low chance of being late on a normal day
        return random.choices(['Late', 'On Time'], weights=[0.05, 0.95])[0]

# Running a small simulation over 10,000 seasonal workdays
total_days = 10000
rainy_days_count = 0
late_when_raining = 0
clear_days_count = 0
late_when_clear = 0

for _ in range(total_days):
    # Let's say it randomly rains on 20% of the days
    raining = random.random() < 0.20
    status = simulate_commute(raining)
    
    if raining:
        rainy_days_count += 1
        if status == 'Late':
            late_when_raining += 1
    else:
        clear_days_count += 1
        if status == 'Late':
            late_when_clear += 1

# Calculating empirical conditional probabilities
prob_late_given_rain = late_when_raining / rainy_days_count if rainy_days_count > 0 else 0
prob_late_given_clear = late_when_clear / clear_days_count if clear_days_count > 0 else 0

print("--- Dependency Simulation Results ---")
print(f"Simulated Data Over {total_days} Days:\n")
print(f"1. On Rainy Days ({rainy_days_count} total): Arrived Late {late_when_raining} times.")
print(f"   Calculated Probability P(Late | Rain): {prob_late_given_rain:.2%}\n")
print(f"2. On Clear Days ({clear_days_count} total): Arrived Late {late_when_clear} times.")
print(f"   Calculated Probability P(Late | Clear): {prob_late_given_clear:.2%}\n")
print("Conclusion: Since P(Late | Rain) != P(Late | Clear), the two events are Dependent!")
```
