```python
import random
```

```python
def simulate_commute(is_raining):
    if is_raining:
        return random.choices(['Late', 'On Time'], weights=[0.60, 0.40])[0]
    else:
        return random.choices(['Late', 'On Time'], weights=[0.05, 0.95])[0]
```

```python
total_days = 10000
rainy_days_count = 0
late_when_raining = 0
clear_days_count = 0
late_when_clear = 0
```

```python
for _ in range(total_days):
    raining = random.random() < 0.20
    status = simulate_commute(raining)

    if raining:
        rainy_days_count += 1
        if status == 'Late':
            late_when_raining += 1
    else:
        clear_days_count += 1
        if status == 'Late':
            late_when_clear += 1
```

```python
p_late_given_rain = late_when_raining / rainy_days_count if rainy_days_count > 0 else 0
p_late_given_clear = late_when_clear / clear_days_count if clear_days_count > 0 else 0

print(p_late_given_rain)
print(p_late_given_clear)
```

---

```python
print(f"Simulated Data Over {total_days} Days:\n")

print(f"1. On Rainy Days ({rainy_days_count} total): Arrived Late {late_when_raining} times.")
print(f"   Calculated Probability P(Late | Rain): {p_late_given_rain:.2%}\n")

print(f"2. On Clear Days ({clear_days_count} total): Arrived Late {late_when_clear} times.")
print(f"   Calculated Probability P(Late | Clear): {p_late_given_clear:.2%}\n")
```
