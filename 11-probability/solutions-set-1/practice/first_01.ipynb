```python
sample_space = {1, 2, 3, 4, 5, 6}
even_numbers = {x for x in sample_space if x % 2 == 0}
```

```python
total_outcomes = len(sample_space)
favorable_outcomes = len(even_numbers)
```

```python
probabitity = favorable_outcomes / total_outcomes
```

```python
print(f"Sample Space : {sample_space}")
print(f"Favorable Outcomes (Even Numbers) : {favorable_outcomes}")
print('-' * 50)
print(f"Probability of rolling an even number : {probabitity:.2f}")
```
