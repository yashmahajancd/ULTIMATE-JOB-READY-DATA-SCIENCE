```python
sample_space = [1, 2, 3, 4, 5, 6]
even_nums = [x for x in sample_space if x % 2 == 0]
```

```python
favorable_outcomes = len(even_nums)
total_outcomes = len(sample_space)
```

```python
probability = favorable_outcomes / total_outcomes
```

---

```python
print(f"Sample space : {sample_space}")
print(f"Favorable outcomes (even numbers) : {even_nums}")
print("-" * 50)
print(f"Probability of rolling an even number : {probability:.2f}")
```
