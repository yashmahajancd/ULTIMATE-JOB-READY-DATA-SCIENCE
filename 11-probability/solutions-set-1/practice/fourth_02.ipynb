```python
bag = {'red': 3, 'blue': 2, 'green': 5}
```

```python
total_balls = sum(bag.values())
```

```python
total_balls
```

```python
red_balls = bag['red']
```

```python
red_balls
```

```python
probability_of_red = red_balls / total_balls
```

```python
probability_of_red
```

---

```python
print(f"Bag contents : {bag}")
print(f"Total balls in the bag : {total_balls}")
print("-" * 60)
print(f"Probability of drawing a red ball : {red_balls}/{total_balls} = {probability_of_red:.2f}")
```
