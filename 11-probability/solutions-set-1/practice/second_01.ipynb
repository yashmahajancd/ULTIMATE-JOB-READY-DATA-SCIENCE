```python
import itertools
```

```python
coins = ['H', 'T']
```

```python
sample_space = list(itertools.product(coins, repeat=2))
```

```python
sample_space
```

```python
sample_space_formatted = ["".join(outcome) for outcome in sample_space]
```

```python
sample_space_formatted
```

---

```python
print("Experiment: Tossing 2 coins")
print("-" * 50)
print(f"Sample Space : {sample_space_formatted}")
print(f"Total Outcomes : {len(sample_space_formatted)}")
```
