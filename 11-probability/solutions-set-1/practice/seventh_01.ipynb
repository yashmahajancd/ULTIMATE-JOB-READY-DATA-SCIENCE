```python
dice_faces = [1, 2, 3, 4, 5, 6]
sample_spaces = [(d1, d2) for d1 in dice_faces for d2 in dice_faces]

print(sample_spaces)
```

```python
favorable_outcomes = [outcome for outcome in sample_spaces if outcome[0] % 2 == 0 and outcome[1] % 2 == 0]

print(favorable_outcomes)
```

```python
total_count = len(sample_spaces)
favorable_count = len(favorable_outcomes)
probability = favorable_count / total_count

print(total_count)
print(favorable_count)
print(probability)
```

---

```python
print(f"Total possible combinations : {total_count}")
print(f"Favorable combinations (both even) : {favorable_count}")
print(f"List of favorable pairs : {favorable_outcomes}")
print("-" * 100)
print(f"Probability : {favorable_count}/{total_count} = {probability:.2f} ({probability * 100:.1f}%)")
```
