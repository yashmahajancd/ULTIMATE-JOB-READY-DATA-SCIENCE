```python
total_cards = 52
red_cards = 26
total_queens = 4
red_queens = 2
```

```python
p_red = red_cards / total_cards
p_queen = total_queens / total_cards
p_red_and_queen = red_queens / total_cards

print(p_red)
print(p_queen)
print(p_red_and_queen)
```

```python
p_red_or_queen = p_red + p_queen - p_red_and_queen

print(p_red_or_queen)
```

---

```python
print(f"Probability of red card : {p_red:.4f}")
print(f"Probability of queen : {p_queen:.4f}")
print(f"Probability of red queen (overlap) : {p_red_and_queen:.4f}")
print("-" * 55)
print(f"Probability of red card or queen : {p_red_or_queen:.4f} (or {p_red_or_queen * 100:.2f}%)")
```
