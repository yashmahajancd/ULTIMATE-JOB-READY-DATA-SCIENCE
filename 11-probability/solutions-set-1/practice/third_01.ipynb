```python
dice = [1, 2, 3, 4, 5, 6]
sample_space = [(d1, d2) for d1 in dice for d2 in dice]
```

```python
favorable_outcomes = [outcome for outcome in sample_space if sum(outcome) == 7]
```

```python
probability = len(favorable_outcomes) / len(sample_space)
```

---

```python
print(f"Total Possible Outcomes : {len(sample_space)}")
print(f"Favorable Outcomes (Sum = 7) : {favorable_outcomes}")
print("-" * 85)
print(f"Probability : {len(favorable_outcomes)}/{len(sample_space)} = {probability:.4f}")
```
