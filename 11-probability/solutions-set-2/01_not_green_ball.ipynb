# 🔴🔵🟢 Question 1: Probability of 'Not' an Event

**Problem:** A box contains 4 red, 3 green, and 2 blue balls. One ball is picked at random. What is the probability that the ball is **not green**?

---

### 🧠 Mathematical Solution

First, let's find the total number of balls in the sample space ($n(S)$):
$$n(S) = 4 \text{ (red)} + 3 \text{ (green)} + 2 \text{ (blue)} = 9 \text{ balls}$$

We need the probability of picking a ball that is **not green**. There are two ways to solve this:

**Method 1: Direct Counting**
If the ball is not green, it must be either red or blue.
$$n(\text{Not Green}) = n(\text{Red}) + n(\text{Blue}) = 4 + 2 = 6$$
$$P(\text{Not Green}) = \frac{6}{9} = \frac{2}{3}$$

**Method 2: Complementary Event**
Calculate the probability of getting a green ball and subtract it from 1.
$$P(\text{Green}) = \frac{3}{9} = \frac{1}{3}$$
$$P(\text{Not Green}) = 1 - P(\text{Green}) = 1 - \frac{1}{3} = \frac{2}{3}$$

> **Result:** The probability that the ball is not green is **2/3** (approx. **0.6667** or 66.67%).

```python
# Define the box contents
box = {
    'Red': 4,
    'Green': 3,
    'Blue': 2
}

# Calculations
total_balls = sum(box.values())
green_balls = box['Green']

# Method 1: Complementary Probability
p_green = green_balls / total_balls
p_not_green = 1 - p_green

# Method 2: Direct Calculation
not_green_balls = total_balls - green_balls
p_not_green_direct = not_green_balls / total_balls

print("--- Box Contents ---")
print(f"Total Balls: {total_balls}")
print(f"Balls that are NOT Green: {not_green_balls}\n")

print(f"Probability (Not Green): {not_green_balls}/{total_balls}")
print(f"Decimal Value: {p_not_green:.4f}")
```
