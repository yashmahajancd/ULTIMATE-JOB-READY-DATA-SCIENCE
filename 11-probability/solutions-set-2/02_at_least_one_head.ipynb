# 🪙 Question 2: At Least One Head

**Problem:** Two coins are tossed. What is the probability of getting **at least one head**?

---

### 🧠 Mathematical Solution

When two coins are tossed, the sample space ($S$) is:
$$S = \{HH, HT, TH, TT\}$$
Total outcomes ($n(S)$) = 4

Let $E$ be the event of getting **at least one head**. This means getting either 1 head or 2 heads.
The favorable outcomes are:
$$E = \{HH, HT, TH\}$$
Number of favorable outcomes ($n(E)$) = 3

The probability $P(E)$ is:
$$P(E) = \frac{n(E)}{n(S)} = \frac{3}{4}$$

> **Result:** The probability of getting at least one head is **3/4** (or **0.75** / 75%).

```python
import itertools

# Define the sample space for 2 coins
coins = ['H', 'T']
sample_space = list(itertools.product(coins, repeat=2))

# Find outcomes with AT LEAST one Head ('H')
favorable_outcomes = [outcome for outcome in sample_space if 'H' in outcome]

# Calculate probability
total_outcomes = len(sample_space)
favorable_count = len(favorable_outcomes)
probability = favorable_count / total_outcomes

# Display nicely formatted outcomes
formatted_space = ["".join(o) for o in sample_space]
formatted_favorable = ["".join(o) for o in favorable_outcomes]

print("--- Coin Toss Simulation ---")
print(f"Sample Space: {formatted_space}")
print(f"Favorable Outcomes (>= 1 Head): {formatted_favorable}\n")
print(f"Probability: {favorable_count}/{total_outcomes} = {probability:.2f}")
```
