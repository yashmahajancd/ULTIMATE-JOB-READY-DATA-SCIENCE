# ⚪⚫ Question 4: Drawing With Replacement

**Problem:** A bag contains 6 white and 4 black balls. Two balls are drawn **with replacement**. What is the probability that **both are white**?

---

### 🧠 Mathematical Solution

"With replacement" means that after the first ball is drawn, it is put back into the bag before the second draw. This means the two draws are **Independent Events**.

Total balls in the bag = $6 \text{ (white)} + 4 \text{ (black)} = 10$.

**Draw 1: Probability of getting a white ball**
$$P(W_1) = \frac{6}{10} = \frac{3}{5}$$

**Draw 2: Probability of getting a white ball (after replacing the first)**
Since the bag's contents are identical to the first draw:
$$P(W_2) = \frac{6}{10} = \frac{3}{5}$$

**Combined Probability (Both White):**
For independent events, we multiply the probabilities:
$$P(W_1 \cap W_2) = P(W_1) \times P(W_2)$$
$$P(\text{Both White}) = \frac{3}{5} \times \frac{3}{5} = \frac{9}{25}$$

> **Result:** The probability that both balls are white is **9/25** (or **0.36** / 36%).

```python
# Bag setup
white_balls = 6
black_balls = 4
total_balls = white_balls + black_balls

# Probability of drawing a white ball on any single draw
p_white_single = white_balls / total_balls

# Since it's WITH REPLACEMENT, the events are independent
# P(White AND White) = P(White) * P(White)
p_both_white = p_white_single * p_white_single

print("----- Drawing With Replacement -----")
print(f"Total Balls: {total_balls} (White: {white_balls}, Black: {black_balls})")
print(f"Probability of White on Draw 1: {p_white_single:.2f}")
print(f"Probability of White on Draw 2: {p_white_single:.2f}")
print("-" * 45)
print(f"Probability of Both White: {p_both_white:.4f} ({p_both_white * 100:.1f}%)")
```
