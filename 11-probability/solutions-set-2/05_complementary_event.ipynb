# 📝 Question 5: Test Question Knowledge

**Problem:** A student knows the answer to 70% of the questions on a test. If one question is selected at random, what is the probability that the student **doesn't know** the answer?

---

### 🧠 Mathematical Solution

This is a classic example of **Complementary Events**. The sum of the probability of an event happening and the probability of it not happening is always 1 (or 100%).

Let $K$ be the event that the student knows the answer.
$$P(K) = 70\% = 0.70$$

Let $K'$ be the event that the student **does not** know the answer.
The formula for a complementary event is:
$$P(K') = 1 - P(K)$$

$$P(K') = 1 - 0.70 = 0.30$$

> **Result:** The probability that the student doesn't know the answer is **0.30** (or **30%**).

```python
# Given probabilities
prob_knows_percentage = 70
prob_knows_decimal = prob_knows_percentage / 100

# Calculating the complementary probability
# P(Doesn't Know) = 1 - P(Knows)
prob_doesnt_know = 1.0 - prob_knows_decimal

print("--- Test Probabilities ---")
print(f"Probability student KNOWS the answer: {prob_knows_decimal:.2f} ({prob_knows_percentage}%)")
print(f"Probability student DOES NOT KNOW the answer: {prob_doesnt_know:.2f} ({prob_doesnt_know * 100:.0f}%)")
```
