```python
p_knows_percentage = 70
p_knows_decimal = p_knows_percentage / 100

print(p_knows_decimal)
```

```python
p_doesnt_know = 1.0 - p_knows_decimal

print(p_doesnt_know)
```

---

```python
print(f"Probability student KNOWS the answer : {p_knows_decimal:.2f} ({p_knows_percentage}%)")
print(f"Probability student DOES NOT KNOW the answer : {p_doesnt_know:.2f} ({p_doesnt_know * 100:.0f}%)")
```
