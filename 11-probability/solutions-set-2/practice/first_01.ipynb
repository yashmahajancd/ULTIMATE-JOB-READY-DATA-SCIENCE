```python
box = {"red": 4, "green": 3, "blue": 2}
```

```python
total_balls = sum(box.values())
green_balls = box['green']

print(total_balls)
print(green_balls)
```

```python
p_green = green_balls / total_balls
p_not_green = 1 - p_green

print(p_green)
print(p_not_green)
```

```python
not_green_balls = total_balls - green_balls
p_not_green_direct = not_green_balls / total_balls

print(not_green_balls)
print(p_not_green_direct)
```

---

```python
print(f"Total balls : {total_balls}")
print(f"Balls that are NOT Green : {not_green_balls}\n")
print(f"Probability (NOT Green) : {not_green_balls}/{total_balls}")
print(f"Decimal value : {p_not_green:.4f}")
```
