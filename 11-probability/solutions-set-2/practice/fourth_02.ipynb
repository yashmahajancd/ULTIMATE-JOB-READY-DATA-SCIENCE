```python
white_balls = 6
black_balls = 4
total_balls = white_balls + black_balls

print(total_balls)
```

```python
p_white_single = white_balls / total_balls

print(p_white_single)
```

```python
p_both_white = p_white_single * p_white_single

print(p_both_white)
```

---

```python
print(f"Total balls: {total_balls} (white: {white_balls}, black: {black_balls})")
print(f"Probability of white on draw 1: {p_white_single:.2f}")
print(f"Probability of white on draw 2: {p_white_single:.2f}")
print("-" * 50)
print(f"Probability of both white: {p_both_white:.4f} ({p_both_white * 100:.1f}%)")
```
