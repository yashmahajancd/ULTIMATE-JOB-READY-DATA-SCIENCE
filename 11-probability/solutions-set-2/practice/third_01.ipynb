```python
sample_space = set(range(1, 11))

print(sample_space)
```

```python
multiples_of_3 = {x for x in sample_space if x % 3 == 0}
multiples_of_5 = {x for x in sample_space if x % 5 == 0}

print(multiples_of_3)
print(multiples_of_5)
```

```python
favorable_outcomes = multiples_of_3.union(multiples_of_5)

print(favorable_outcomes)
```

```python
probability = len(favorable_outcomes) / len(sample_space)

print(probability)
```

---

```python
print(f"Sample space : {sample_space}")
print(f"Multiples of 3 : {multiples_of_3}")
print(f"Multiples of 5 : {multiples_of_5}")
print(f"Combined favorable outcomes (union) : {favorable_outcomes}\n")
print(f"Probability : {len(favorable_outcomes)}/{len(sample_space)} = {probability:.2f}")
```
