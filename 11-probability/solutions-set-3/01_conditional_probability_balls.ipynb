# 🎱 Question 1: Conditional Probability (Ball Analogy)

**Problem:** Suppose you have a box containing 5 red balls, 3 blue balls, and 2 green balls. Now, you’re told that the ball picked is **not green**. What is the probability that it’s red?

---

### 🧠 Mathematical Solution

The formula for Conditional Probability is defined as:
$$P(A | B) = \frac{P(A \text{ and } B)}{P(B)}$$

Let us define the events:
* **Event A:** The ball is red.
* **Event B:** The ball is not green.

#### 🔢 Step-by-Step Calculation:
1. **Total number of balls** = $5 \text{ (Red)} + 3 \text{ (Blue)} + 2 \text{ (Green)} = 10$ balls.
2. **Number of balls that are NOT green ($n(B)$)** = $5 \text{ (Red)} + 3 \text{ (Blue)} = 8$ balls.
3. **Number of balls that are both Red AND Not Green ($n(A \cap B)$)** = $5$ (Since all red balls are automatically classified as 'not green').

Now, calculating the probability based on the updated sample space:
$$P(\text{Red} | \text{Not Green}) = \frac{\text{Favorable Outcomes}}{\text{Total Remaining Outcomes}}$$

$$P(\text{Red} | \text{Not Green}) = \frac{5}{8}$$

> **Result:** If the picked ball is not green, the probability of it being red is **5/8** (which is **0.625** or **62.5%**).

```python
# Box ke andar ki balls define karte hain
balls = {
    'Red': 5,
    'Blue': 3,
    'Green': 2
}

# Total sample space size
total_balls = sum(balls.values())

# Event B: Ball is NOT Green
favorable_B = balls['Red'] + balls['Blue']  # 5 + 3 = 8

# Event A and B: Ball is Red AND NOT Green
favorable_A_and_B = balls['Red']  # 5

# Conditional Probability calculation: P(A|B) = n(A and B) / n(B)
prob_red_given_not_green = favorable_A_and_B / favorable_B

print("--- 🎱 Ball Probability Simulation ---")
print(f"Total Balls in Box: {total_balls}")
print(f"Balls that are NOT Green: {favorable_B}")
print(f"Balls that are Red and NOT Green: {favorable_A_and_B}")
print("-" * 40)
print(f"P(Red | Not Green) = {favorable_A_and_B}/{favorable_B} = {prob_red_given_not_green:.3f}")
```
