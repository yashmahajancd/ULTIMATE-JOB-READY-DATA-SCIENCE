# 🎯 Question 2: Law of Total Probability & Bayes' Application

**Problem:** A class has 60% boys and 40% girls. 30% of the boys and 10% of the girls play football. If a randomly selected student is a football player, what is the probability that the student is a boy?

---

### 🧠 Mathematical Solution

Let us express the given values using conditional probability notation:
* $P(B)$ (Student is a boy) = $60\% = 0.6$
* $P(G)$ (Student is a girl) = $40\% = 0.4$
* $P(F | B)$ (Plays football given they are a boy) = $30\% = 0.3$
* $P(F | G)$ (Plays football given they are a girl) = $10\% = 0.1$

---

### 📊 Step 1: Calculate the Total Probability of Playing Football $P(F)$
According to the Law of Total Probability:
$$P(F) = P(F | B) \times P(B) + P(F | G) \times P(G)$$

$$P(F) = (0.3 \times 0.6) + (0.1 \times 0.4)$$
$$P(F) = 0.18 + 0.04 = 0.22$$

---

### 📐 Step 2: Apply the Conditional Probability Formula
We need to find $P(B | F)$, which represents the probability that the student is a boy given that they play football:
$$P(B | F) = \frac{P(F | B) \times P(B)}{P(F)}$$

$$P(B | F) = \frac{0.18}{0.22} \approx 0.818$$

> **Result:** If a randomly selected student is a football player, the probability that the student is a boy is approximately **0.818** (or **81.8%**).

```python
# Given data points
p_boy = 0.6
p_girl = 0.4

p_football_given_boy = 0.3
p_football_given_girl = 0.1

# Step 1: Total Probability of playing football P(F)
p_football = (p_football_given_boy * p_boy) + (p_football_given_girl * p_girl)

# Step 2: Conditional Probability P(Boy | Football)
p_boy_given_football = (p_football_given_boy * p_boy) / p_football

print("--- 📊 Football Player Analysis ---")
print(f"Total Probability of a student playing Football P(F): {p_football:.2f}")
print(f"P(Boy | Football) = {p_football_given_boy * p_boy:.2f} / {p_football:.2f}")
print(f"Result: {p_boy_given_football:.4f} (yaani {p_boy_given_football * 100:.1f}%)")
```
