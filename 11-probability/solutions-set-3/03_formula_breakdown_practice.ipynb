# 🔍 Question 3: Formula Breakdown & Common Mistakes

**Problem:** If the probability of both events $A$ and $B$ happening is $P(A \text{ and } B) = 0.15$, and the probability of event $B$ happening is $P(B) = 0.50$. Calculate the conditional probability $P(A | B)$. Also, identify the common mistakes to avoid in this core concept.

---

### 🧠 Mathematical Solution

The objective of conditional probability is to determine: "Among all the outcomes where $B$ occurs, how many also include $A$?"

Formula:
$$P(A | B) = \frac{P(A \text{ and } B)}{P(B)}$$

Substituting the given values:
$$P(A | B) = \frac{0.15}{0.50} = 0.30$$

---

### ⚠️ Common Mistakes to Avoid

Based on foundational theory, never make these 3 mistakes when working with conditional probabilities:
1. **Confusing $P(A | B)$ with $P(B | A)$:** $P(A|B)$ signifies the probability of A occurring given that B has already happened, whereas $P(B|A)$ implies the opposite scenario.
2. **Forgetting to divide by $P(B)$:** The denominator must always represent the probability of the condition that has already taken place.
3. **Assuming $P(A | B) == P(B | A)$:** These two values are completely distinct and are generally not equal.

> **Result:** $P(A | B) = 0.30$ (or **30%**).

```python
# Inputs according to the formula breakdown
p_A_and_B = 0.15
p_B = 0.50

# Applying conditional probability formula
p_A_given_B = p_A_and_B / p_B

print("--- 🔍 Conditional Probability Evaluation ---")
print(f"P(A and B) = {p_A_and_B}")
print(f"P(B) = {p_B}")
print("-" * 45)
print(f"Calculated P(A | B) = {p_A_given_B:.2f}")

# Warning check logic in code
print("\n🚨 Reminder of Mistakes to Avoid:")
print("- Do NOT assume P(A|B) is equal to P(B|A).")
print("- Always ensure you divide by the correct condition probability (P(B)).")
```
