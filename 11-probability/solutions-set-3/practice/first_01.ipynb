```python
balls = {"red": 5, "blue": 3, "green": 2}
```

```python
total_balls = sum(balls.values())

print(total_balls)
```

```python
favorable_B = balls['red'] + balls['blue']

print(favorable_B)
```

```python
favorable_A_and_B = balls['red']

print(favorable_A_and_B)
```

```python
p_red_given_not_green = favorable_A_and_B / favorable_B

print(p_red_given_not_green)
```

---

```python
print(f"Total balls in box : {total_balls}")
print(f"Balls that are not green : {favorable_B}")
print(f"Balls that are red and not green : {favorable_A_and_B}")
print("-" * 40)
print(f"P(Red | Not Green) : {favorable_A_and_B}/{favorable_B} = {p_red_given_not_green:.3f}")
```
