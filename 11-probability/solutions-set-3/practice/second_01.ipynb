```python
p_boy = 0.6
p_girl = 0.4
```

```python
p_football_given_boy = 0.3
p_football_given_girl = 0.1
```

```python
p_football = (p_football_given_boy * p_boy) + (p_football_given_girl * p_girl)

print(p_football)
```

```python
p_boy_given_football = (p_football_given_boy * p_boy) / p_football

print(p_boy_given_football)
```

---

```python
print(f"Total probability of a student playing football P(F) : {p_football:.2f}")
print(f"P(Boy | Football) : {p_football_given_boy * p_boy:.2f} / {p_football:.2f}")
print("-" * 65)
print(f"Result : {p_boy_given_football:.3f} (or {p_boy_given_football * 100:.1f}%)")
```
