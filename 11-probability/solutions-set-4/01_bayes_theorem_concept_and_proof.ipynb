# 🎯 Question 1: Concept and Mathematical Proof of Bayes' Theorem

**Problem:** State Bayes' Theorem, explain the meaning of each term in the formula, and provide a formal proof deriving it from the fundamental definition of conditional probability.

---

### 🧠 Mathematical Explanation & Proof

Bayes’ Theorem allows us to update the probability of an event based on new evidence or information. 

#### 1. The Standard Formula
The mathematical equation for Bayes' Theorem is expressed as:
$$P(A | B) = \frac{P(B | A) \times P(A)}{P(B)}$$

Where each term represents:
* **$P(A | B)$ (Posterior Probability):** The probability of event $A$ occurring given that event $B$ has already occurred.
* **$P(B | A)$ (Likelihood):** The probability of event $B$ occurring given that event $A$ is true.
* **$P(A)$ (Prior Probability):** The baseline or initial probability of event $A$ before looking at any evidence.
* **$P(B)$ (Marginal Probability):** The total probability of event $B$ occurring across all scenarios.

#### 2. The Expanded Form
Using the Law of Total Probability, the denominator $P(B)$ can be expanded into a weighted sum of all possible conditional branches:
$$P(A | B) = \frac{P(B | A) \times P(A)}{P(B | A) \times P(A) + P(B | \text{No } A) \times P(\text{No } A)}$$

---

### 📝 Step-by-Step Derivation

We can derive this directly from the fundamental definition of conditional probability:

1. **Conditional Probability of $A$ given $B$:**
   $$P(A | B) = \frac{P(A \text{ and } B)}{P(B)} \quad \text{--- (Equation 1)}$$

2. **Conditional Probability of $B$ given $A$:**
   $$P(B | A) = \frac{P(A \text{ and } B)}{P(A)} \quad \text{--- (Equation 2)}$$

3. **Isolating the Joint Probability ($P(A \text{ and } B)$):**
   By multiplying both sides of Equation 2 by $P(A)$, we get:
   $$P(A \text{ and } B) = P(B | A) \times P(A) \quad \text{--- (Equation 3)}$$

4. **Final Substitution:**
   Now, substitute the value of $P(A \text{ and } B)$ from Equation 3 back into the numerator of Equation 1:
   $$P(A | B) = \frac{P(B | A) \times P(A)}{P(B)}$$

This completes the proof of **Bayes' Theorem**.

```python
def compute_bayes_theorem(p_prior_A, p_likelihood_B_given_A, p_likelihood_B_given_not_A):
    """
    Computes the posterior probability P(A|B) using the expanded form of Bayes' Theorem.
    """
    # Calculate P(Not A)
    p_prior_not_A = 1 - p_prior_A
    
    # Calculate total marginal probability P(B)
    p_marginal_B = (p_likelihood_B_given_A * p_prior_A) + (p_likelihood_B_given_not_A * p_prior_not_A)
    
    # Calculate posterior probability P(A|B)
    p_posterior_A_given_B = (p_likelihood_B_given_A * p_prior_A) / p_marginal_B
    
    return p_posterior_A_given_B, p_marginal_B

# Example evaluation using arbitrary baseline numbers
prior_A = 0.30
likelihood_true = 0.80
likelihood_false = 0.10

posterior, marginal = compute_bayes_theorem(prior_A, likelihood_true, likelihood_false)

print("--- 🎯 Bayes' Theorem General Proof Validation ---")
print(f"Given Prior P(A) = {prior_A}")
print(f"Given Likelihood P(B | A) = {likelihood_true}")
print(f"Given False Positive Rate P(B | Not A) = {likelihood_false}")
print("-" * 60)
print(f"Calculated Total Marginal Probability P(B) = {marginal:.4f}")
print(f"Calculated Updated Posterior Probability P(A | B) = {posterior:.4f}")
```
