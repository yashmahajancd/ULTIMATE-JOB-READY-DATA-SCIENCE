# 🏥 Question 2: Medical Diagnostics and False Positives

**Problem:** Suppose a specific disease affects **1% of the population**. A diagnostic clinical test has the following parameters:
* If a person **has** the disease, the test returns a **positive** result 99% of the time.
* If a person **does not have** the disease, the test returns a **positive** result 5% of the time.

If a randomly selected individual takes the test and obtains a **positive** result, what is the actual probability that they truly have the disease?

---

### 🧠 Mathematical Solution

Let us map out the problem using the variables of Bayes' Theorem:
* **$A$ (Disease):** The patient has the disease.
* **$B$ (Positive):** The laboratory test is positive.

#### 🔢 Given Statistical Data:
* Baseline rate of disease: $P(\text{Disease}) = 0.01$
* Baseline rate of health: $P(\text{No Disease}) = 1 - 0.01 = 0.99$
* Sensitivity (True Positive Rate): $P(\text{Positive} | \text{Disease}) = 0.99$
* False Positive Rate: $P(\text{Positive} | \text{No Disease}) = 0.05$

---

### 📊 Step 1: Calculate the Total Probability of a Positive Test $P(\text{Positive})$

Using the Law of Total Probability, we compute all possible paths that generate a positive test result:
$$P(\text{Positive}) = P(\text{Positive} | \text{Disease}) \times P(\text{Disease}) + P(\text{Positive} | \text{No Disease}) \times P(\text{No Disease})$$

$$P(\text{Positive}) = (0.99 \times 0.01) + (0.05 \times 0.99)$$
$$P(\text{Positive}) = 0.0099 + 0.0495 = 0.0594$$

---

### 📐 Step 2: Apply Bayes' Theorem to Find $P(\text{Disease} | \text{Positive})$

Now, evaluate the probability that the patient actually has the disease out of all positive test cases:
$$P(\text{Disease} | \text{Positive}) = \frac{P(\text{Positive} | \text{Disease}) \times P(\text{Disease})}{P(\text{Positive})}$$

$$P(\text{Disease} | \text{Positive}) = \frac{0.0099}{0.0594} \approx 0.1667$$

---

### ✅ Conclusion & Insight

Even though the test claims a high **99% accuracy rate**, the probability that a person actually has the disease given a positive result is only **16.67%**.

**Why does this happen?** Because the base disease rate is exceptionally rare (only 1% of the population). The absolute number of false positives coming from the healthy 99% of the population heavily outweighs the true positive results coming from the small sick population.

```python
# Input values based on medical trial specifications
p_disease = 0.01
p_no_disease = 0.99

p_positive_given_disease = 0.99
p_positive_given_no_disease = 0.05

# Step 1: Computing Total Probability of testing positive P(Positive)
p_total_positive = (p_positive_given_disease * p_disease) + (p_positive_given_no_disease * p_no_disease)

# Step 2: Evaluating Bayes' Theorem for P(Disease | Positive)
p_disease_given_positive = (p_positive_given_disease * p_disease) / p_total_positive

print("--- 🏥 Diagnostic Test Evaluation ---")
print(f"Probability of a Positive Test Result P(Positive): {p_total_positive:.4f}")
print(f"Formula Setup: ({p_positive_given_disease} * {p_disease}) / {p_total_positive:.4f}")
print("-" * 60)
print(f"Probability of actually having the disease: {p_disease_given_positive:.4f}")
print(f"Percentage Value: {p_disease_given_positive * 100:.2f}%")
```
