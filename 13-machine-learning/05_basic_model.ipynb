```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [150, 0],
    [170, 0],
    [130, 1],
    [120, 1]
]
```

```python
labels = ["Apple", "Apple", "Orange", "Orange"]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

```python
prediction = clf.predict([[150, 1], [34, 0]])
```

```python
print(f"The first fruit is {prediction[0]}")
print(f"The second fruit is {prediction[1]}")
```

 ---

```python
print(clf.predict([[160, 0]]))
print(clf.predict([[115, 1]]))
```

## Visualize It

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8, 5))
plot_tree(clf, feature_names=["weight", "color"], class_names=["Apple", "Orange"], filled=True)
plt.tight_layout()
plt.show()
```
