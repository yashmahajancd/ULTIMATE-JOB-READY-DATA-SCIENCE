```python
from sklearn.ensemble import RandomForestClassifier
```

```python
features = [
    [2, 45],
    [3, 55],
    [4, 60],
    [5, 70],
    [6, 80],
    [8, 95]
]
```

```python
labels = [
    "Fail",
    "Fail",
    "Pass",
    "Pass",
    "Pass",
    "Pass"
]
```

```python
clf = RandomForestClassifier()
```

```python
clf = clf.fit(features, labels)
```

---

```python
prediction = clf.predict([
    [7, 90],
    [1, 90],
    [9, 10],
])
```

```python
print(f"Student 1 will {prediction[0]}")
print(f"Student 2 will {prediction[1]}")
print(f"Student 3 will {prediction[2]}")
```
