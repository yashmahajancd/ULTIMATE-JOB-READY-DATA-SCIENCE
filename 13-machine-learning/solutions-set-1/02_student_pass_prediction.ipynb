```python
from sklearn import tree
```

```python
features = [
    [2, 45],
    [3, 55],
    [4, 60],
    [5, 70],
    [6, 80],
    [8, 95]
]
```

```python
labels = [
    "Fail",
    "Fail",
    "Pass",
    "Pass",
    "Pass",
    "Pass"
]
```

```python
clf = tree.DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

---

```python
print(clf.predict([[7, 90]]))
print(clf.predict([[1, 90]]))
```
