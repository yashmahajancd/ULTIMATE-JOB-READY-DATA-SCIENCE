```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [35, 90],
    [30, 80],
    [28, 70],
    [25, 60],
    [20, 50],
    [18, 45],
]
```

```python
labels = [
    "No",
    "No",
    "Yes",
    "Yes",
    "Yes",
    "Yes"
]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

---

```python
print(clf.predict([[27, 65]]))
print(clf.predict([[44, 35]]))
```
