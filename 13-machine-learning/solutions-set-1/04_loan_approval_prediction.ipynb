```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [20000, 500],
    [25000, 550],
    [30000, 580],
    [50000, 650],
    [70000, 720],
    [90000, 800]
]
```

```python
labels = [
    "Rejected",
    "Rejected",
    "Rejected",
    "Approved",
    "Approved",
    "Approved"
]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

---

```python
print(clf.predict([[70000, 720]]))
print(clf.predict([[40000, 615]]))
print(clf.predict([[80000, 555]]))
```
