```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [4, 30],
    [5, 32],
    [6, 35],
    [20, 60],
    [25, 70],
    [30, 75]
]
```

```python
labels = [
    "Cat",
    "Cat",
    "Cat",
    "Dog",
    "Dog",
    "Dog"
]
```

```python
clf = DecisionTreeClassifier()
```

```python
clf = clf.fit(features, labels)
```

---

```python
print(clf.predict([[18, 55]]))
print(clf.predict([[13, 47]]))
print(clf.predict([[10, 50]]))
```
