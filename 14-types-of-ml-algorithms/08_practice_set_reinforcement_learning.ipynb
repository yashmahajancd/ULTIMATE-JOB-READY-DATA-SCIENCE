# 1️⃣ Robot Path Learning Using Q-Learning

Build a **Q-Learning** agent that learns the optimal path for a robot to reach its goal.

States:

- 0 → Start
- 1 → Middle
- 2 → Near Goal
- 3 → Goal

Rewards:

- Reaching the Goal = **+100**
- All other valid moves = **0**

Train the agent for **100 episodes** and display the learned **Q-table**.
