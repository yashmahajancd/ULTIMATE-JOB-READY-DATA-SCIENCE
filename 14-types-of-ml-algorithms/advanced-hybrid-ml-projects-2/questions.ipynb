# Advanced Hybrid Machine Learning – Problem Bank (Q11–Q15)

> **Domain Overview:** Each problem below combines **Unsupervised Learning (Clustering)** for pattern discovery with **Supervised Learning (Classification/Regression)** for predictive modeling across critical real-world industries.

## Q11. Predictive Maintenance & Industrial Machine Failure Detection ⭐⭐⭐⭐⭐

* **Domain:** Manufacturing & Industrial IoT
* **Problem Statement:** A manufacturing company wants to identify machines that may fail before an actual breakdown occurs, while also discovering distinct machine operating patterns.
* **Target Variable:** `Machine Failure` (`0` = No Failure, `1` = Failure)

### Dataset Features
| Feature | Unit / Type | Description |
| :--- | :--- | :--- |
| **Machine Temperature** | Continuous (°C) | Operating thermal reading |
| **Operating Hours** | Continuous (hrs) | Total cumulative runtime |
| **Vibration Level** | Continuous (mm/s) | Mechanical oscillation intensity |
| **Rotational Speed** | Continuous (RPM) | Spindle/motor revolutions per minute |
| **Pressure Level** | Continuous (bar/psi) | Internal hydraulic/pneumatic pressure |
| **Maintenance Frequency** | Discrete (count) | Scheduled servicing intervals completed |
| **Machine Age** | Continuous (years) | Time since installation |
| **Power Consumption** | Continuous (kW) | Electrical load during operation |

### Implementation Tasks
1. **Data Preprocessing:** Handle missing values and outliers.
2. **Normalization:** Perform appropriate feature scaling.
3. **Pattern Discovery:** Apply **K-Means Clustering** to identify different machine operating patterns.
4. **Feature Augmentation:** Add the resulting cluster labels as a new feature to the dataset.
5. **Predictive Modeling:** Train a **Random Forest Classifier** to predict machine failure.
6. **Model Evaluation:** Evaluate performance using **Precision, Recall, F1-Score, Confusion Matrix, and ROC-AUC**.
7. **Feature Importance:** Identify the most critical machine failure indicators.
8. **Inference:** Predict the failure probability for a new, unseen machine record.
9. **Business Impact:** Explain how this hybrid system helps a company reduce unexpected maintenance costs and downtime.

## Q12. E-Commerce Customer Lifetime Value & Revenue Prediction ⭐⭐⭐⭐⭐

* **Domain:** Retail & E-Commerce Analytics
* **Problem Statement:** An e-commerce company wants to segment its customer base into distinct behavioral groups and predict how much revenue each customer will generate over the next six months.
* **Target Variable:** `Future Revenue` (₹)

### Dataset Features
| Feature | Unit / Type | Description |
| :--- | :--- | :--- |
| **Customer Age** | Discrete (years) | Age of the registered user |
| **Total Orders** | Discrete (count) | Lifetime completed purchases |
| **Average Order Value** | Continuous (₹) | Mean expenditure per transaction |
| **Purchase Frequency** | Continuous (orders/mo) | Rate of repeat buying |
| **Total Spending** | Continuous (₹) | Cumulative historical expenditure |
| **Website Visit Frequency** | Continuous (visits/mo) | Platform engagement rate |
| **Cart Abandonment Rate** | Percentage (%) | Ratio of unpurchased carts to total carts |
| **Discount Usage** | Percentage / Count | Reliance on promo codes/offers |
| **Customer Membership Duration** | Continuous (months) | Tenure since account creation |

### Implementation Tasks
1. **Data Preprocessing:** Handle missing values and skewed numerical distributions.
2. **Feature Engineering:** Create derived metrics using customer purchase history (e.g., spend velocity, engagement-to-purchase ratio).
3. **Customer Segmentation:** Apply **K-Means Clustering** to group customers into distinct purchasing tiers.
4. **Feature Augmentation:** Add customer cluster labels to the dataset.
5. **Predictive Modeling:** Train an **XGBoost Regressor** or **Random Forest Regressor** to predict six-month future revenue.
6. **Model Evaluation:** Evaluate performance using **MAE, RMSE, and R² Score**.
7. **Feature Importance:** Identify the most influential revenue-generating features.
8. **Inference:** Predict six-month revenue for a new customer profile.
9. **Business Strategy:** Recommend tailored marketing strategies for each discovered customer segment.

## Q13. Cybersecurity – Network Intrusion & Anomaly Detection ⭐⭐⭐⭐⭐

* **Domain:** Network Security & Threat Intelligence
* **Problem Statement:** A cybersecurity firm wants to isolate suspicious network activities and classify whether incoming network traffic represents a cyberattack.
* **Target Variable:** `Network Attack` (`0` = Normal Traffic, `1` = Cyberattack)

### Dataset Features
| Feature | Unit / Type | Description |
| :--- | :--- | :--- |
| **Network Packet Size** | Continuous (bytes) | Size of transmitted data packets |
| **Connection Duration** | Continuous (sec) | Length of the active network session |
| **Source Bytes** | Continuous (bytes) | Data volume sent from source to destination |
| **Destination Bytes** | Continuous (bytes) | Data volume sent from destination to source |
| **Failed Login Attempts** | Discrete (count) | Consecutive authentication failures |
| **Connection Frequency** | Continuous (conn/sec) | Rate of incoming connection requests |
| **Protocol Type** | Categorical | Communication protocol (e.g., TCP, UDP, ICMP) |
| **Port Number** | Discrete / Categorical | Target network port (e.g., 80, 443, 22) |
| **Traffic Rate** | Continuous (Mbps) | Bandwidth throughput rate |

### Implementation Tasks
1. **Data Preprocessing:** Handle missing values and encode categorical features (e.g., Protocol Type).
2. **Normalization:** Apply appropriate feature scaling.
3. **Anomaly Grouping:** Use **K-Means Clustering** to separate baseline normal traffic from unusual network patterns.
4. **Feature Augmentation:** Append cluster labels/distances to the dataset.
5. **Predictive Modeling:** Train an **Isolation Forest** or **Random Forest-based** detection pipeline.
6. **Model Evaluation:** Evaluate performance using **Precision, Recall, F1-Score, and ROC-AUC**.
7. **Behavioral Analysis:** Analyze characteristics of unusual network activity clusters.
8. **Inference:** Detect suspicious activity from a new incoming network traffic sample.
9. **Security Impact:** Explain how this hybrid pipeline strengthens real-time cyberattack detection.

## Q14. Healthcare – Hospital Readmission Prediction & Patient Segmentation ⭐⭐⭐⭐⭐

* **Domain:** Clinical Analytics & Healthcare Operations
* **Problem Statement:** A hospital wants to identify patients at high risk of returning within 30 days of discharge, while segmenting patient populations to optimize care planning.
* **Target Variable:** `30-Day Readmission` (`0` = No, `1` = Yes)

### Dataset Features
| Feature | Unit / Type | Description |
| :--- | :--- | :--- |
| **Patient Age** | Discrete (years) | Age at time of admission |
| **Previous Hospital Admissions** | Discrete (count) | Historical inpatient stays |
| **Length of Hospital Stay** | Discrete (days) | Duration of current inpatient admission |
| **Number of Diagnoses** | Discrete (count) | Total comorbidities/conditions recorded |
| **Medication Count** | Discrete (count) | Number of active prescriptions at discharge |
| **Blood Pressure** | Continuous (mmHg) | Systolic/diastolic clinical reading |
| **Glucose Level** | Continuous (mg/dL) | Blood sugar concentration |
| **Previous Emergency Visits** | Discrete (count) | Prior ER visits within the past year |
| **Treatment Cost** | Continuous (₹) | Total billed inpatient care expense |

### Implementation Tasks
1. **Data Preprocessing:** Handle missing clinical values and extreme outliers.
2. **Normalization:** Perform feature scaling across physiological and utilization metrics.
3. **Patient Stratification:** Apply **K-Means Clustering** to identify distinct patient risk/care profiles.
4. **Feature Augmentation:** Add patient cluster labels to the dataset.
5. **Predictive Modeling:** Train a **Gradient Boosting Classifier** to predict 30-day readmission risk.
6. **Model Evaluation:** Evaluate performance using **Recall, Precision, F1-Score, ROC-AUC, and Confusion Matrix**.
7. **Risk Factor Analysis:** Identify the primary clinical and utilization factors driving readmissions.
8. **Inference:** Predict readmission probability for a newly discharged patient.
9. **Clinical Application:** Explain how hospital care teams can use these predictions to design targeted post-discharge follow-up plans.

## Q15. Banking – Customer Default Risk & Financial Behaviour Analysis ⭐⭐⭐⭐⭐

* **Domain:** FinTech, Credit Risk & Retail Banking
* **Problem Statement:** A bank wants to segment borrowers by financial behavior patterns and predict which applicants are likely to default on loan repayments.
* **Target Variable:** `Loan Default` (`0` = No Default, `1` = Default)

### Dataset Features
| Feature | Unit / Type | Description |
| :--- | :--- | :--- |
| **Customer Age** | Discrete (years) | Age of the loan applicant |
| **Monthly Income** | Continuous (₹) | Verified net monthly earnings |
| **Credit Score** | Discrete (300–900) | Bureau creditworthiness rating |
| **Loan Amount** | Continuous (₹) | Principal amount requested |
| **Existing Debt** | Continuous (₹) | Total active liabilities |
| **Monthly EMI** | Continuous (₹) | Scheduled monthly installment obligation |
| **Employment Duration** | Continuous (years) | Tenure at current employer |
| **Previous Loan Defaults** | Discrete (count) | Historical delinquency record |
| **Credit Utilization Ratio** | Percentage (%) | Ratio of used credit to total credit limit |
| **Payment History** | Categorical / Score | Track record of on-time vs. delayed payments |

### Implementation Tasks
1. **Data Preprocessing:** Handle missing values and financial outliers.
2. **Feature Engineering:** Compute key financial health indicators such as **Debt-to-Income (DTI) ratio** and **EMI-to-Income ratio**.
3. **Behavioral Segmentation:** Apply **K-Means Clustering** to group customers based on financial stability and borrowing habits.
4. **Feature Augmentation:** Add customer cluster labels to the dataset.
5. **Predictive Modeling:** Train an **XGBoost Classifier** or **Random Forest Classifier** to predict default risk.
6. **Model Evaluation:** Evaluate performance using **Precision, Recall, F1-Score, ROC-AUC, and Confusion Matrix**.
7. **Risk Drivers:** Identify the major financial factors influencing loan default.
8. **Inference:** Predict default probability for a new loan applicant.
9. **Governance & Compliance:** Explain how banks can leverage this model for responsible, transparent credit-risk assessment.
