```python
# Step 1: Create Dataset

data = pd.DataFrame({
    "MathScore": [
        90, 85, 88,
        60, 65, 70,
        50, 45, 55,
        75, 80, 78
    ],
    "ProgrammingScore": [
        95, 90, 92,
        85, 80, 88,
        40, 35, 45,
        65, 70, 68
    ],
    "CommunicationScore": [
        60, 65, 62,
        75, 80, 78,
        90, 95, 88,
        85, 82, 80
    ],
    "CreativityScore": [
        55, 60, 58,
        65, 70, 68,
        95, 90, 92,
        80, 78, 82
    ],
    "ProblemSolvingScore": [
        92, 88, 90,
        72, 75, 78,
        50, 45, 55,
        80, 82, 85
    ],
    "Career": [
        "Data Scientist",
        "Data Scientist",
        "Data Scientist",

        "Web Developer",
        "Web Developer",
        "Web Developer",

        "UI/UX Designer",
        "UI/UX Designer",
        "UI/UX Designer",

        "Business Analyst",
        "Business Analyst",
        "Business Analyst"
    ]
})

print("\nOriginal Dataset:\n")

data
```

---

```python
# Step 2: Features and Target

X = data.drop("Career", axis=1)

y = data["Career"]
```

---

```python
# Step 3: Standardize Data

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

---

```python
# Step 4: K-Means Clustering

kmeans = KMeans(n_clusters=4, random_state=42)

clusters = kmeans.fit_predict(X_scaled)

data["Cluster"] = clusters

print("\nClusters:\n")

print(data[["Career", "Cluster"]])
```
