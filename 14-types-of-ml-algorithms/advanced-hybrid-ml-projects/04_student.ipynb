```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestClassifier
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix
)
```

---

```python
# Step 1: Create Dataset

data = pd.DataFrame({
    "MathScore": [
        90, 85, 88,
        60, 65, 70,
        50, 45, 55,
        75, 80, 78
    ],
    "ProgrammingScore": [
        95, 90, 92,
        85, 80, 88,
        40, 35, 45,
        65, 70, 68
    ],
    "CommunicationScore": [
        60, 65, 62,
        75, 80, 78,
        90, 95, 88,
        85, 82, 80
    ],
    "CreativityScore": [
        55, 60, 58,
        65, 70, 68,
        95, 90, 92,
        80, 78, 82
    ],
    "ProblemSolvingScore": [
        92, 88, 90,
        72, 75, 78,
        50, 45, 55,
        80, 82, 85
    ],
    "Career": [
        "Data Scientist",
        "Data Scientist",
        "Data Scientist",

        "Web Developer",
        "Web Developer",
        "Web Developer",

        "UI/UX Designer",
        "UI/UX Designer",
        "UI/UX Designer",

        "Business Analyst",
        "Business Analyst",
        "Business Analyst"
    ]
})

print("\nOriginal Dataset:\n")

data
```

---

```python
# Step 2: Features and Target

X = data.drop("Career", axis=1)

y = data["Career"]
```

---

```python
# Step 3: Standardize Data

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

---

```python
# Step 4: K-Means Clustering

kmeans = KMeans(n_clusters=4, random_state=42)

clusters = kmeans.fit_predict(X_scaled)

data["Cluster"] = clusters

print("\nClusters:\n")

print(data[["Career", "Cluster"]])
```

---

```python
# Step 5: Add Cluster as New Feature

X_final = pd.DataFrame(X_scaled, columns=X.columns)

X_final["Cluster"] = clusters
```

---

```python
# Step 6: Split Dataset

X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.25, random_state=42)
```

---

```python
# Step 7: Train Random Forest

model = RandomForestClassifier(n_estimators=100, random_state=42)
    
model = model.fit(X_train, y_train)
```

---

```python
# Step 8: Prediction

y_pred = model.predict(X_test)
```

---

```python
# Step 9: Accuracy

accuracy = accuracy_score(y_test, y_pred)

print(f"\nAccuracy: {accuracy}")
```

---

```python
# Step 10: Confusion Matrix

cm = confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:\n")

print(cm)
```

---

```python
# Step 11: Predict New Student Career

new_student = pd.DataFrame({
    "MathScore": [87],
    "ProgrammingScore": [91],
    "CommunicationScore": [65],
    "CreativityScore": [60],
    "ProblemSolvingScore": [90]
})
```

```python
# Standardize

new_student_scaled = scaler.transform(new_student)
```

```python
# Find cluster

new_cluster = kmeans.predict(new_student_scaled)
```

```python
# Create final input

new_student_final = pd.DataFrame(new_student_scaled, columns=X.columns)

new_student_final["Cluster"] = new_cluster
```

```python
# Predict career

prediction = model.predict(new_student_final)

print("\nPredicted Career:\n")

print(prediction[0])
```
