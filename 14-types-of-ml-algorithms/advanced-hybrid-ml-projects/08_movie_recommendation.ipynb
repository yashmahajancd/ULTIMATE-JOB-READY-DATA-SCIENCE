```python
import pandas as pd
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.ensemble import RandomForestRegressor
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

```python
import numpy as np
```

---

```python
# Step 1: Create Dataset

data = pd.DataFrame({

    "UserAge": [
        18, 20, 22, 25, 28,
        30, 32, 35, 38, 40,
        42, 45, 48, 50, 55,
        60, 65, 24, 33, 52
    ],
    "MoviesWatched": [
        5, 8, 10, 15, 18,
        20, 25, 28, 30, 35,
        38, 40, 45, 50, 55,
        60, 65, 12, 32, 48
    ],
    "AverageWatchTime": [
        45, 50, 55, 60, 65,
        70, 75, 80, 82, 85,
        88, 90, 92, 95, 100,
        105, 110, 58, 84, 98
    ],
    "PreferredGenreScore": [
        40, 45, 50, 55, 60,
        62, 65, 68, 70, 72,
        75, 78, 80, 82, 85,
        88, 90, 52, 70, 84
    ],
    "PreviousRatings": [
        2.5, 2.7, 2.8, 3.0, 3.1,
        3.2, 3.3, 3.4, 3.5, 3.6,
        3.7, 3.8, 4.0, 4.1, 4.2,
        4.4, 4.5, 2.9, 3.5, 4.1
    ],
    "MovieRating": [
        2.5, 2.7, 2.8, 3.0, 3.1,
        3.2, 3.4, 3.5, 3.6, 3.7,
        3.8, 4.0, 4.1, 4.2, 4.3,
        4.5, 4.7, 2.9, 3.6, 4.2
    ]
})
```

```python
print("\nOriginal Dataset:\n")
data
```

---

```python
# Step 2: Separate Features and Target

X = data.drop("MovieRating", axis=1)

y = data["MovieRating"]
```

---

```python
# Step 3: Standardize Features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

---

```python
# Step 4: User Segmentation using K-Means

kmeans = KMeans(n_clusters=5, random_state=42, n_init=10)

clusters = kmeans.fit_predict(X_scaled)
```

```python
# Add cluster to dataset

data["ViewingCluster"] = clusters

print("\nUser Viewing Clusters:\n")

print(data[["UserAge", "MoviesWatched", "ViewingCluster"]])
```

---

```python
# Step 5: Add Cluster as a New Feature

X_final = pd.DataFrame(X_scaled, columns=["UserAge", "MoviesWatched", "AverageWatchTime", "PreferredGenreScore", "PreviousRatings"])

X_final["ViewingCluster"] = clusters
```

---

```python
# Step 6: Train-Test Split

X_train, X_test, y_train, y_test = train_test_split(X_final, y, test_size=0.25, random_state=42)
```

---

```python
# Step 7: Train Random Forest Regressor

model = RandomForestRegressor(n_estimators=100, random_state=42)

model = model.fit(X_train, y_train)
```

---

```python
# Step 8: Predict Ratings

y_pred = model.predict(X_test)
```
