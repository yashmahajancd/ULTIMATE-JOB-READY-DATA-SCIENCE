```python
import numpy as np
```

```python
R = np.array([
    [-1, 0, -1, -1],
    [-1, -1, 0, -1],
    [-1, -1, -1, 100],
    [-1, -1, -1, 100]
])

Q = np.zeros((4,4))

gamma = 0.8
```

```python
for episode in range(100):

    state = np.random.randint(0,4)

    while state != 3:

        actions = np.where(R[state] != -1)[0]

        action = np.random.choice(actions)

        next_state = action

        Q[state, action] = R[state, action] + gamma * np.max(Q[next_state])

        state = next_state
```

```python
print(Q)
```
