```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
# features: [Income, Credit Score]
customers = [
    [20000, 500],
    [25000, 550],
    [30000, 600],
    [70000, 750],
    [80000, 800],
    [90000, 850]
]

loan_status = [
    "Rejected",
    "Rejected",
    "Rejected",
    "Approved",
    "Approved",
    "Approved"
]
```

```python
# unsupervised Learning
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans = kmeans.fit(customers)
```

```python
# supervised Learning
model = DecisionTreeClassifier()
model = model.fit(customers, loan_status)

prediction = model.predict([[75000, 780]])
```

---

```python
print(f"Clusters : {kmeans.labels_}")

print(f"Loan Status : {prediction}")
```
