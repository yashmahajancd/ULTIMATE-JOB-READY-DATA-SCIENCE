```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
members = [
    [18, 55],
    [20, 60],
    [22, 65],
    [35, 80],
    [40, 85],
    [45, 90]
]

membership = [
    "Basic",
    "Basic",
    "Basic",
    "Premium",
    "Premium",
    "Premium"
]
```

```python
# unsupervised
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(members)
```

```python
# supervised
model = DecisionTreeClassifier()
model = model.fit(members, membership)

prediction = model.predict([[25, 70]])
```

---

```python
print(f"Clusters : {kmeans.labels_}")

print(f"Membership : {prediction}")
```
