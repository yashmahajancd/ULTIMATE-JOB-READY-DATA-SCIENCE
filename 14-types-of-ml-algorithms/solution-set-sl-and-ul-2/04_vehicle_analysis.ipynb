```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
vehicles = [
    [60, 50],
    [70, 45],
    [80, 40],
    [120, 20],
    [140, 15],
    [160, 12]
]

vehicle_type = [
    "Bike",
    "Bike",
    "Bike",
    "Sports Car",
    "Sports Car",
    "Sports Car"
]
```

```python
# unsupervised
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(vehicles)
```

```python
# supervised
model = DecisionTreeClassifier()
model = model.fit(vehicles, vehicle_type)

prediction = model.predict([[130, 18]])
```

---

```python
print(f"Clusters : {kmeans.labels_}")

print(f"Vehicle Type : {prediction}")
```
