```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
students = [
    [6.5, 55],
    [7.0, 60],
    [7.5, 65],
    [8.5, 85],
    [9.0, 90],
    [9.2, 95]
]

admission = [
    "Rejected",
    "Rejected",
    "Rejected",
    "Admitted",
    "Admitted",
    "Admitted"
]
```

```python
# unsupervised
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(students)
```

```python
# supervised
model = DecisionTreeClassifier()
model = model.fit(students, admission)

prediction = model.predict([[8.8, 88]])
```

---

```python
print(f"Clusters : {kmeans.labels_}")

print(f"Admission Status : {prediction}")
```
