```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.DataFrame({
    "MonthlyCharges":[30,35,40,70,75,80],
    "Tenure":[2,3,5,20,25,30],
    "Churn":[1,1,1,0,0,0]
})
```

```python
data
```

---

```python
X = data[["MonthlyCharges","Tenure"]]
y = data["Churn"]
```

---

```python
# unsupervised
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans = kmeans.fit(X)

data["Cluster"] = kmeans.labels_
```

```python
data
```

---

```python
# supervised
model = DecisionTreeClassifier()
model = model.fit(X,y)

# prediction = model.predict([[65,18]])
prediction_data = pd.DataFrame([[65, 18]], columns=['MonthlyCharges', 'Tenure'])  # changed to match training features
prediction = model.predict(prediction_data)

print(f"Prediction: {prediction}")
```
