```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.DataFrame({
    "Experience":[1,2,3,8,10,12],
    "Score":[55,60,65,90,92,95],
    "Promotion":["No","No","No","Yes","Yes","Yes"]
})
```

```python
data
```

---

```python
X = data[["Experience","Score"]]
y = data["Promotion"]
```

---

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(X)

data["Cluster"]=kmeans.labels_
```

```python
data
```

---

```python
model = DecisionTreeClassifier()
# model = model.fit(X,y)
model = model.fit(X.values, y.values)

prediction = model.predict([[7,88]])

print(f"Prediction: {prediction}")
```
