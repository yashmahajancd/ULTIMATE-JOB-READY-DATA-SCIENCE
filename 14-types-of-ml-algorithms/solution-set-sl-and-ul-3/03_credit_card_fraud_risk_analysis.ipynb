```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.DataFrame({
    "Amount":[200,300,400,8000,9000,12000],
    "Frequency":[2,3,5,15,18,20],
    "Fraud":[0,0,0,1,1,1]
})
```

```python
data
```

---

```python
X=data[["Amount","Frequency"]]
y=data["Fraud"]
```

---

```python
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans = kmeans.fit(X)

data["Risk"]=kmeans.labels_
```

```python
data
```

---

```python
model = DecisionTreeClassifier()
model = model.fit(X.values, y.values)

prediction = model.predict([[9500,17]])

print(f"Prediction: {prediction}")
```
