```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
data = pd.DataFrame({
    "Time":[5,6,8,30,35,40],
    "Purchase":[500,700,900,6000,7000,9000],
    "BuyAgain":[0,0,0,1,1,1]
})
```

```python
data
```

---

```python
X = data[["Time","Purchase"]]
y = data["BuyAgain"]
```

---

```python
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(X)

data["Cluster"]=kmeans.labels_
```

```python
data
```

---

```python
model = DecisionTreeClassifier()
model = model.fit(X.values, y.values)

prediction = model.predict([[32,6500]])

print(f"Prediction: {prediction}")
```
