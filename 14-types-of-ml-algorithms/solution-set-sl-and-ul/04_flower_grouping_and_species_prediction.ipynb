```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
from sklearn.tree import DecisionTreeClassifier
```

```python
flowers = [
    [1.4, 0.2],
    [1.5, 0.3],
    [5.0, 1.8],
    [5.2, 2.0]
]

species = [
    "Setosa",
    "Setosa",
    "Versicolor",
    "Versicolor"
]
```

```python
# unsupervised
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans = kmeans.fit(flowers)
```

```python
# supervised
model = DecisionTreeClassifier()
model = model.fit(flowers, species)
```

---

```python
print(kmeans.labels_)

print(model.predict([[1.6, 0.2]]))
```
