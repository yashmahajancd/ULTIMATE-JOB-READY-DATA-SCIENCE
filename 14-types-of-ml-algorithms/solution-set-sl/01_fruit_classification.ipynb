```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [150, 0],
    [170, 0],
    [130, 1],
    [120, 1]
]
```

```python
labels = [
    "Apple",
    "Apple",
    "Orange",
    "Orange"
]
```

```python
model = DecisionTreeClassifier()
```

```python
model = model.fit(features, labels)
```

---

```python
print(model.predict([[160, 0]]))
print(model.predict([[115, 1]]))
```
