```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [5,80],
    [6,90],
    [1,40],
    [2,50],
    [7,95]
]
```

```python
labels = [
    "Pass",
    "Pass",
    "Fail",
    "Fail",
    "Pass"
]
```

```python
model = DecisionTreeClassifier()
```

```python
model = model.fit(features, labels)
```

---

```python
print(model.predict([[4,85]]))
print(model.predict([[1,45]]))
```
