```python
from sklearn.linear_model import LinearRegression
```

```python
experience = [
    [1],
    [2],
    [3],
    [4],
    [5]
]
```

```python
salary = [
    25000,
    35000,
    45000,
    55000,
    65000
]
```

```python
model = LinearRegression()
```

```python
model = model.fit(experience, salary)
```

```python
prediction = model.predict([[6]])
```
