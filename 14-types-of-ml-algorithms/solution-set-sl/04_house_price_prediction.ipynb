```python
from sklearn.linear_model import LinearRegression
```

```python
area = [
    [1000],
    [1200],
    [1500],
    [1800],
    [2000]
]
```

```python
price = [
    3000000,
    3600000,
    4500000,
    5400000,
    6000000
]
```

```python
model = LinearRegression()
```

```python
model = model.fit(area, price)
```

---

```python
print(model.predict([[1700]]))
```
