```python
from sklearn.tree import DecisionTreeClassifier
```

```python
features = [
    [10],
    [15],
    [1],
    [2],
    [20],
    [3]
]
```

```python
labels = [
    "Spam",
    "Spam",
    "Not Spam",
    "Not Spam",
    "Spam",
    "Not Spam"
]
```

```python
model = DecisionTreeClassifier()
```

```python
model = model.fit(features, labels)
```

---

```python
print(model.predict([[12]]))
print(model.predict([[2]]))
```
