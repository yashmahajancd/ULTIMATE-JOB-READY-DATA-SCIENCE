```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
```

```python
customers = [
    [20, 30000],
    [22, 35000],
    [25, 40000],
    [45, 80000],
    [48, 85000],
    [50, 90000]
]
```

```python
model = KMeans(n_clusters=3, random_state=42)
```

```python
model = model.fit(customers)
```

---

```python
print(model.labels_)
```
