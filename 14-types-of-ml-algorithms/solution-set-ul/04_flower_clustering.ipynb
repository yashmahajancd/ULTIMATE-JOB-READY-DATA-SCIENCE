```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
```

```python
flowers = [
    [1.4, 0.2],
    [1.5, 0.3],
    [5.0, 1.8],
    [5.2, 2.0],
    [4.8, 1.7]
]
```

```python
model = KMeans(n_clusters=2, random_state=42)
```

```python
model = model.fit(flowers)
```

---

```python
print(model.labels_)
```
