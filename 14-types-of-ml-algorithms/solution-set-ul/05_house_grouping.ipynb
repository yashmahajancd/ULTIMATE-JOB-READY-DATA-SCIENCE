```python
import os
os.environ['OMP_NUM_THREADS'] = '1'
```

```python
from sklearn.cluster import KMeans
```

```python
houses = [
    [900, 3000000],
    [1000, 3500000],
    [1100, 3700000],
    [2000, 8000000],
    [2200, 9000000],
    [2500, 9500000]
]
```

```python
model = KMeans(n_clusters=2, random_state=42)
```

```python
model = model.fit(houses)
```

---

```python
print(model.labels_)
```
