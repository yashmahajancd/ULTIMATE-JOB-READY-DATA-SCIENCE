## 🧹 Handling Missing Data

Some features, like <mark>`total_bedrooms`</mark>, contain missing values. You can:

1. Drop rows with missing values

2. Drop the entire column

3. **Impute missing values** (recommended)

We’ll use **option 3** using <mark>`SimpleImputer`</mark> from Scikit-Learn, which allows consistent handling across all datasets (train, test, new data):

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy="median")
housing_num = housing.select_dtypes(include=[np.number])

imputer.fit(housing_num)
```

This computes the median for each numerical column and stores it in <mark>`imputer.statistics_`</mark>:

```python
>>> imputer.statistics_
array([-118.51, 34.26, 29., 2125., 434., 1167., 408., 3.5385])
```

Now apply the learned medians to transform the data:

```python
X = imputer.transform(housing_num)
```
