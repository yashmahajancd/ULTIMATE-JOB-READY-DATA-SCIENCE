# ⚙️ Scikit-Learn Design Principles

Scikit-Learn has a simple and consistent API that makes it easy to use and understand. Below are the key design principles behind it:

---

## 📌 1. Consistency

All objects follow a standard interface, which makes learning and using different tools in Scikit-Learn easier.

---

## 🧠 2. Estimators

Any object that learns from data is called an **estimator**.

- Use the <mark>`.fit()`</mark> method to train an estimator.

- In supervised learning, pass both `X` (features) and `y` (labels) to <mark>`.fit(X, y)`</mark>.

- Hyperparameters (like <mark>`strategy='mean'`</mark> in <mark>`SimpleImputer`</mark>) are set when creating the object.

### Example:

```python
imputer = SimpleImputer(strategy="median")
imputer.fit(data)
```

---

## 🔄 3. Transformers

Some estimators can also transform data. These are called **transformers**.

- Use <mark>`.transform()`</mark> to apply the transformation after fitting.

- Use <mark>`.fit_transform()`</mark> to do both in one step.

### Example:

```python
X_transformed = imputer.fit_transform(data)
```

---

## 🔮 4. Predictors

Models that can make predictions are **predictors**.

- Use <mark>`.predict()`</mark> to make predictions on new data.

- Use <mark>`.score()`</mark> to evaluate performance (e.g., accuracy or R²).

### Example:

```python
model = LinearRegression()
model.fit(X_train, y_train)
predictions = model.predict(X_test)
score = model.score(X_test, y_test)
```

---

## 🔍 5. Inspection

- Hyperparameters can be accessed directly: <mark>`model.param_name`</mark>

- Learned parameters are stored with an underscore: <mark>`model.coef_`</mark>, <mark>`imputer.statistics_`</mark>

---

## 📦 6. No Extra Classes

- Inputs and outputs are basic structures like NumPy arrays or Pandas DataFrames.

- No need to learn custom data types.

---

## 🏗️ 7. Composition

You can combine steps into a **Pipeline**, chaining transformers and a final predictor.

### Example:

```python
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("model", LinearRegression())
])

pipeline.fit(X, y)
```

---

## 🚀 8. Sensible Defaults

Most tools in Scikit-Learn work well with default settings, so you can get started quickly.

---
